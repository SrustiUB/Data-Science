```python
# ============================================================
# NIFTY GAP-UP / GAP-DOWN PROBABILITY
# AutoML + QuantStats
# ============================================================

!pip -q install yfinance flaml quantstats

import yfinance as yf
import pandas as pd
import numpy as np
import quantstats as qs

from flaml import AutoML
from sklearn.metrics import roc_auc_score, log_loss, brier_score_loss

# 1. DATA
df = yf.download(
    "^NSEI",
    start="2010-01-01",
    auto_adjust=False,
    progress=False
)

if isinstance(df.columns, pd.MultiIndex):
    df.columns = df.columns.get_level_values(0)

df = df[["Open","High","Low","Close","Volume"]].copy()

# Today's gap = today's open vs yesterday's close
df["Gap"] = df["Open"] / df["Close"].shift(1) - 1

df["Return"] = df["Close"].pct_change()

# ------------------------------------------------------------
# 2. FEATURES
# ------------------------------------------------------------

for n in [1,3,5,10,20]:
    df[f"Ret_{n}"] = df["Close"].pct_change(n)

for n in [5,10,20]:
    df[f"Vol_{n}"] = df["Return"].rolling(n).std()

df["Range"] = (
    df["High"] - df["Low"]
) / df["Close"]

df["SMA20"] = df["Close"].rolling(20).mean()

df["Trend"] = (
    df["Close"] / df["SMA20"] - 1
)

df["PrevGap"] = df["Gap"].shift(1)

df["GapVol20"] = df["Gap"].rolling(20).std()

# ------------------------------------------------------------
# 3. TARGETS
# ------------------------------------------------------------

# NEXT day's gap
df["Gap_Up"] = (
    df["Gap"].shift(-1) > 0.01
).astype(int)

df["Gap_Down"] = (
    df["Gap"].shift(-1) < -0.01
).astype(int)

features = [
    "Ret_1","Ret_3","Ret_5","Ret_10","Ret_20",
    "Vol_5","Vol_10","Vol_20",
    "Range","Trend","PrevGap","GapVol20"
]

df = df.dropna(subset=features)

X = df[features]

# ------------------------------------------------------------
# 4. TRAIN / TEST
# ------------------------------------------------------------

split = int(len(df)*0.80)

X_train = X.iloc[:split]
X_test = X.iloc[split:]

# ------------------------------------------------------------
# 5. GAP-UP AUTOML
# ------------------------------------------------------------

y_up = df["Gap_Up"]

up_train = y_up.iloc[:split]
up_test = y_up.iloc[split:]

model_up = AutoML()

model_up.fit(
    X_train,
    up_train,
    task="classification",
    metric="log_loss",
    time_budget=60,
    seed=42,
    eval_method="holdout"
)

p_up = model_up.predict_proba(X_test)[:,1]

# ------------------------------------------------------------
# 6. GAP-DOWN AUTOML
# ------------------------------------------------------------

y_down = df["Gap_Down"]

down_train = y_down.iloc[:split]
down_test = y_down.iloc[split:]

model_down = AutoML()

model_down.fit(
    X_train,
    down_train,
    task="classification",
    metric="log_loss",
    time_budget=60,
    seed=42,
    eval_method="holdout"
)

p_down = model_down.predict_proba(X_test)[:,1]

# ------------------------------------------------------------
# 7. EVALUATION
# ------------------------------------------------------------

print("\nGAP-UP MODEL")
print("------------")

print("ROC-AUC:",
      round(roc_auc_score(up_test,p_up),4))

print("Log Loss:",
      round(log_loss(up_test,p_up),4))

print("Brier:",
      round(brier_score_loss(up_test,p_up),4))

print("\nGAP-DOWN MODEL")
print("--------------")

print("ROC-AUC:",
      round(roc_auc_score(down_test,p_down),4))

print("Log Loss:",
      round(log_loss(down_test,p_down),4))

print("Brier:",
      round(brier_score_loss(down_test,p_down),4))

# ------------------------------------------------------------
# 8. PROBABILITY DASHBOARD
# ------------------------------------------------------------

test = df.iloc[split:].copy()

test["P_GapUp"] = p_up
test["P_GapDown"] = p_down

print("\nLatest Probabilities")
print("--------------------")

print(
    "P(Gap Up > 1%):",
    round(p_up[-1]*100,2),
    "%"
)

print(
    "P(Gap Down < -1%):",
    round(p_down[-1]*100,2),
    "%"
)

# ------------------------------------------------------------
# 9. OVERNIGHT GAP STRATEGY
# ------------------------------------------------------------

# Signal is generated at previous close.
# Today's gap is therefore tradable at the open.

test["Position"] = np.where(
    test["P_GapUp"] >= 0.60,
    1,
    np.where(
        test["P_GapDown"] >= 0.60,
        -1,
        0
    )
)

test["Position"] = test["Position"].shift(1).fillna(0)

test["Strategy_Return"] = (
    test["Position"] * test["Gap"]
)

# ------------------------------------------------------------
# 10. QUANTSTATS
# ------------------------------------------------------------

strategy = test["Strategy_Return"]

print("\nQuantStats — Gap Strategy")
print("--------------------------")

print(
    "Sharpe:",
    round(qs.stats.sharpe(strategy),3)
)

print(
    "CAGR:",
    round(qs.stats.cagr(strategy)*100,2),
    "%"
)

print(
    "Max Drawdown:",
    round(
        qs.stats.max_drawdown(strategy)*100,
        2
    ),
    "%"
)

print(
    "Volatility:",
    round(
        qs.stats.volatility(strategy)*100,
        2
    ),
    "%"
)

qs.reports.basic(strategy)

print("\nProject 9 completed.")
```
