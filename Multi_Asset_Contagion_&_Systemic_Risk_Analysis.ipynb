```python
# ================================================================
# PART 1: MULTI-ASSET DATA & DEPENDENCY NETWORK
# Financial Contagion & Systemic Risk Project
# ================================================================

# Install required libraries
!pip -q install yfinance networkx plotly

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import yfinance as yf
import networkx as nx
import matplotlib.pyplot as plt
import plotly.graph_objects as go

print("=" * 70)
print("PART 1: MULTI-ASSET DATA & DEPENDENCY NETWORK")
print("=" * 70)

# ------------------------------------------------
# 1. ASSET UNIVERSE
# ------------------------------------------------

assets = {
    "BTC": "BTC-USD",
    "ETH": "ETH-USD",
    "BNB": "BNB-USD",
    "SOL": "SOL-USD",
    "XRP": "XRP-USD",
    "ADA": "ADA-USD",
    "DOGE": "DOGE-USD",
    "AVAX": "AVAX-USD",
    "LINK": "LINK-USD",
    "LTC": "LTC-USD"
}

tickers = list(assets.values())

print("\nAssets:")
print(list(assets.keys()))

# ------------------------------------------------
# 2. DOWNLOAD DATA
# ------------------------------------------------

print("\nDownloading market data...")

data = yf.download(
    tickers,
    period="2y",
    interval="1d",
    auto_adjust=True,
    progress=False
)

# Extract closing prices
prices = data["Close"].copy()

# Rename columns
prices.columns = [
    symbol
    for symbol in assets.keys()
]

# Remove completely missing observations
prices = prices.dropna(how="all")

# Forward fill small gaps
prices = prices.ffill()

print("\n" + "=" * 70)
print("PRICE DATA")
print("=" * 70)

print("Shape:", prices.shape)

display(prices.head())

# ------------------------------------------------
# 3. DATA QUALITY
# ------------------------------------------------

print("\n" + "=" * 70)
print("DATA QUALITY")
print("=" * 70)

quality = pd.DataFrame({
    "asset": prices.columns,
    "observations": [
        prices[col].notna().sum()
        for col in prices.columns
    ],
    "missing": [
        prices[col].isna().sum()
        for col in prices.columns
    ],
    "start": [
        prices[col].first_valid_index()
        for col in prices.columns
    ],
    "end": [
        prices[col].last_valid_index()
        for col in prices.columns
    ]
})

display(quality)

# ------------------------------------------------
# 4. LOG RETURNS
# ------------------------------------------------

returns = np.log(
    prices / prices.shift(1)
)

returns = returns.replace(
    [np.inf, -np.inf],
    np.nan
)

returns = returns.dropna()

print("\n" + "=" * 70)
print("RETURN DATA")
print("=" * 70)

print("Return observations:", len(returns))

display(returns.head())

# ------------------------------------------------
# 5. RETURN STATISTICS
# ------------------------------------------------

statistics = pd.DataFrame({
    "mean_daily_return":
        returns.mean(),

    "daily_volatility":
        returns.std(),

    "annualized_return":
        returns.mean() * 365,

    "annualized_volatility":
        returns.std() * np.sqrt(365),

    "minimum_return":
        returns.min(),

    "maximum_return":
        returns.max()
})

print("\n" + "=" * 70)
print("ASSET RETURN STATISTICS")
print("=" * 70)

display(
    statistics.sort_values(
        "annualized_volatility",
        ascending=False
    )
)

# ------------------------------------------------
# 6. CORRELATION MATRIX
# ------------------------------------------------

correlation = returns.corr()

print("\n" + "=" * 70)
print("RETURN CORRELATION MATRIX")
print("=" * 70)

display(
    correlation.round(3)
)

# ------------------------------------------------
# 7. ROLLING CORRELATION
# ------------------------------------------------

rolling_window = 60

rolling_corr_btc = pd.DataFrame(
    index=returns.index
)

for asset in returns.columns:

    rolling_corr_btc[asset] = (
        returns["BTC"]
        .rolling(rolling_window)
        .corr(returns[asset])
    )

print("\n" + "=" * 70)
print("60-DAY ROLLING CORRELATION WITH BTC")
print("=" * 70)

display(
    rolling_corr_btc.tail()
)

# ------------------------------------------------
# 8. BUILD DEPENDENCY NETWORK
# ------------------------------------------------

# Threshold for meaningful dependency
threshold = 0.50

G = nx.Graph()

# Add nodes
for asset in returns.columns:
    G.add_node(asset)

# Add edges where absolute correlation
# exceeds threshold

for i in correlation.index:

    for j in correlation.columns:

        if i >= j:
            continue

        corr_value = correlation.loc[i, j]

        if abs(corr_value) >= threshold:

            G.add_edge(
                i,
                j,
                weight=abs(corr_value),
                correlation=corr_value
            )

print("\n" + "=" * 70)
print("DEPENDENCY NETWORK")
print("=" * 70)

print("Nodes:", G.number_of_nodes())
print("Edges:", G.number_of_edges())

# ------------------------------------------------
# 9. NETWORK CENTRALITY
# ------------------------------------------------

degree_centrality = nx.degree_centrality(G)

weighted_degree = dict(
    G.degree(weight="weight")
)

betweenness = nx.betweenness_centrality(
    G,
    weight="weight"
)

centrality = pd.DataFrame({
    "asset":
        list(G.nodes()),

    "degree_centrality":
        [
            degree_centrality[x]
            for x in G.nodes()
        ],

    "weighted_degree":
        [
            weighted_degree[x]
            for x in G.nodes()
        ],

    "betweenness_centrality":
        [
            betweenness[x]
            for x in G.nodes()
        ]
})

centrality = centrality.sort_values(
    "weighted_degree",
    ascending=False
)

print("\n" + "=" * 70)
print("NETWORK CENTRALITY")
print("=" * 70)

display(
    centrality.round(4)
)

# ------------------------------------------------
# 10. MOST CONNECTED ASSETS
# ------------------------------------------------

print("\n" + "=" * 70)
print("MOST CONNECTED ASSETS")
print("=" * 70)

for _, row in centrality.head(5).iterrows():

    print(
        f"{row['asset']}: "
        f"weighted connectivity = "
        f"{row['weighted_degree']:.3f}"
    )

# ------------------------------------------------
# 11. NETWORK VISUALIZATION
# ------------------------------------------------

plt.figure(
    figsize=(12, 8)
)

pos = nx.spring_layout(
    G,
    seed=42,
    weight="weight"
)

node_sizes = [
    800 + weighted_degree[node] * 1500
    for node in G.nodes()
]

edge_widths = [
    G[u][v]["weight"] * 4
    for u, v in G.edges()
]

nx.draw_networkx_nodes(
    G,
    pos,
    node_size=node_sizes,
    alpha=0.85
)

nx.draw_networkx_edges(
    G,
    pos,
    width=edge_widths,
    alpha=0.5
)

nx.draw_networkx_labels(
    G,
    pos,
    font_size=11,
    font_weight="bold"
)

plt.title(
    "Cryptocurrency Dependency Network"
)

plt.axis("off")

plt.show()

# ------------------------------------------------
# 12. STRONGEST DEPENDENCIES
# ------------------------------------------------

edges = []

for u, v, attributes in G.edges(
    data=True
):

    edges.append({
        "asset_1": u,
        "asset_2": v,
        "correlation":
            attributes["correlation"],
        "absolute_correlation":
            attributes["weight"]
    })

strongest_edges = pd.DataFrame(
    edges
)

if len(strongest_edges) > 0:

    strongest_edges = (
        strongest_edges
        .sort_values(
            "absolute_correlation",
            ascending=False
        )
    )

print("\n" + "=" * 70)
print("STRONGEST ASSET DEPENDENCIES")
print("=" * 70)

display(
    strongest_edges.head(15)
)

# ------------------------------------------------
# 13. SAVE OUTPUTS
# ------------------------------------------------

prices.to_csv(
    "/content/P1_MultiAsset_Prices.csv"
)

returns.to_csv(
    "/content/P1_MultiAsset_Returns.csv"
)

correlation.to_csv(
    "/content/P1_Correlation_Matrix.csv"
)

statistics.to_csv(
    "/content/P1_Return_Statistics.csv"
)

centrality.to_csv(
    "/content/P1_Network_Centrality.csv",
    index=False
)

if len(strongest_edges) > 0:

    strongest_edges.to_csv(
        "/content/P1_Strongest_Dependencies.csv",
        index=False
    )

# ------------------------------------------------
# 14. FINAL SUMMARY
# ------------------------------------------------

print("\n" + "=" * 70)
print("PART 1 COMPLETE")
print("=" * 70)

print("""
PROJECT:
Financial Contagion Network

PART 1 ACHIEVEMENTS:

✓ Multi-asset market data collected
✓ Log returns calculated
✓ Return statistics generated
✓ Correlation structure measured
✓ Rolling BTC correlations calculated
✓ Dependency network constructed
✓ Network centrality calculated
✓ Strongest asset relationships identified

NEXT:
PART 2 → SPILLOVER & CONTAGION ANALYSIS
""")
```

```python
# ================================================================
# PART 2: SPILLOVER & CONTAGION ANALYSIS
# Financial Contagion Network
# ================================================================

!pip -q install statsmodels networkx

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import networkx as nx

from statsmodels.tsa.stattools import grangercausalitytests

print("=" * 70)
print("PART 2: SPILLOVER & CONTAGION ANALYSIS")
print("=" * 70)

# ------------------------------------------------
# 1. LOAD PART 1 RETURNS
# ------------------------------------------------

returns_file = "/content/P1_MultiAsset_Returns.csv"

returns = pd.read_csv(
    returns_file,
    index_col=0,
    parse_dates=True
)

returns = returns.dropna()

print("\nLoaded returns:", returns.shape)

# ------------------------------------------------
# 2. GRANGER CAUSALITY
# ------------------------------------------------

print("\n" + "=" * 70)
print("GRANGER CAUSALITY ANALYSIS")
print("=" * 70)

assets = list(returns.columns)

max_lag = 3

spillover_results = []

for source in assets:

    for target in assets:

        if source == target:
            continue

        pair = returns[
            [target, source]
        ].dropna()

        try:

            result = grangercausalitytests(
                pair,
                maxlag=max_lag,
                verbose=False
            )

            # Take the minimum p-value
            # across tested lags

            p_values = []

            for lag in range(
                1,
                max_lag + 1
            ):

                p = result[
                    lag
                ][0]["ssr_ftest"][1]

                p_values.append(p)

            min_p = min(p_values)

            best_lag = (
                np.argmin(p_values) + 1
            )

            significant = (
                min_p < 0.05
            )

            spillover_results.append({

                "source": source,

                "target": target,

                "min_p_value": min_p,

                "best_lag": best_lag,

                "significant":
                    significant
            })

        except Exception:

            spillover_results.append({

                "source": source,

                "target": target,

                "min_p_value": np.nan,

                "best_lag": np.nan,

                "significant": False
            })


spillover = pd.DataFrame(
    spillover_results
)

# ------------------------------------------------
# 3. SIGNIFICANT SPILLOVER RELATIONSHIPS
# ------------------------------------------------

significant_spillovers = (
    spillover[
        spillover["significant"] == True
    ]
    .sort_values(
        "min_p_value"
    )
)

print(
    "\nSignificant spillover relationships:",
    len(significant_spillovers)
)

display(
    significant_spillovers.head(30)
)

# ------------------------------------------------
# 4. BUILD DIRECTED CONTAGION NETWORK
# ------------------------------------------------

G = nx.DiGraph()

for asset in assets:

    G.add_node(asset)

for _, row in significant_spillovers.iterrows():

    G.add_edge(
        row["source"],
        row["target"],
        weight=-np.log10(
            max(
                row["min_p_value"],
                1e-10
            )
        ),
        p_value=row["min_p_value"],
        lag=row["best_lag"]
    )

print("\n" + "=" * 70)
print("DIRECTED CONTAGION NETWORK")
print("=" * 70)

print(
    "Nodes:",
    G.number_of_nodes()
)

print(
    "Directed spillover edges:",
    G.number_of_edges()
)

# ------------------------------------------------
# 5. RISK TRANSMITTER SCORE
# ------------------------------------------------

out_degree = dict(
    G.out_degree(
        weight="weight"
    )
)

in_degree = dict(
    G.in_degree(
        weight="weight"
    )
)

# ------------------------------------------------
# 6. OUT / IN STRENGTH
# ------------------------------------------------

transmitter_score = pd.Series(
    out_degree
)

receiver_score = pd.Series(
    in_degree
)

network_results = pd.DataFrame({

    "asset": assets,

    "risk_transmitter_score":
        [
            transmitter_score.get(
                x,
                0
            )
            for x in assets
        ],

    "risk_receiver_score":
        [
            receiver_score.get(
                x,
                0
            )
            for x in assets
        ]
})

# ------------------------------------------------
# 7. NET CONTAGION SCORE
# ------------------------------------------------

network_results[
    "net_contagion_score"
] = (

    network_results[
        "risk_transmitter_score"
    ]

    -

    network_results[
        "risk_receiver_score"
    ]
)

network_results = (
    network_results
    .sort_values(
        "net_contagion_score",
        ascending=False
    )
)

print("\n" + "=" * 70)
print("RISK TRANSMITTER / RECEIVER ANALYSIS")
print("=" * 70)

display(
    network_results.round(4)
)

# ------------------------------------------------
# 8. CLASSIFY ASSETS
# ------------------------------------------------

def classify_asset(row):

    net = row[
        "net_contagion_score"
    ]

    if net > 1:

        return "NET RISK TRANSMITTER"

    elif net < -1:

        return "NET RISK RECEIVER"

    else:

        return "BALANCED"


network_results[
    "contagion_role"
] = network_results.apply(
    classify_asset,
    axis=1
)

print("\n" + "=" * 70)
print("CONTAGION ROLE")
print("=" * 70)

display(
    network_results
)

# ------------------------------------------------
# 9. TOP RISK TRANSMITTERS
# ------------------------------------------------

print("\n" + "=" * 70)
print("TOP RISK TRANSMITTERS")
print("=" * 70)

top_transmitters = (
    network_results
    .sort_values(
        "risk_transmitter_score",
        ascending=False
    )
)

display(
    top_transmitters.head(5)
)

# ------------------------------------------------
# 10. TOP RISK RECEIVERS
# ------------------------------------------------

print("\n" + "=" * 70)
print("TOP RISK RECEIVERS")
print("=" * 70)

top_receivers = (
    network_results
    .sort_values(
        "risk_receiver_score",
        ascending=False
    )
)

display(
    top_receivers.head(5)
)

# ------------------------------------------------
# 11. STRONGEST SPILLOVER CHANNELS
# ------------------------------------------------

print("\n" + "=" * 70)
print("STRONGEST SPILLOVER CHANNELS")
print("=" * 70)

display(
    significant_spillovers[
        [
            "source",
            "target",
            "min_p_value",
            "best_lag"
        ]
    ].head(20)
)

# ------------------------------------------------
# 12. NETWORK CENTRALITY
# ------------------------------------------------

if G.number_of_edges() > 0:

    pagerank = nx.pagerank(
        G,
        weight="weight"
    )

    network_results[
        "contagion_pagerank"
    ] = [
        pagerank.get(
            asset,
            0
        )
        for asset in assets
    ]

else:

    network_results[
        "contagion_pagerank"
    ] = 0

print("\n" + "=" * 70)
print("SYSTEMIC CONNECTIVITY")
print("=" * 70)

display(
    network_results.sort_values(
        "contagion_pagerank",
        ascending=False
    ).round(4)
)

# ------------------------------------------------
# 13. SAVE RESULTS
# ------------------------------------------------

spillover.to_csv(
    "/content/P2_All_Spillover_Tests.csv",
    index=False
)

significant_spillovers.to_csv(
    "/content/P2_Significant_Spillovers.csv",
    index=False
)

network_results.to_csv(
    "/content/P2_Contagion_Network_Ranking.csv",
    index=False
)

# ------------------------------------------------
# 14. FINAL INTERPRETATION
# ------------------------------------------------

print("\n" + "=" * 70)
print("PART 2 INTERPRETATION")
print("=" * 70)

print("""
The analysis now moves beyond simple correlation.

A correlation tells us that two assets move together.

Granger causality asks whether past movements in one asset
contain information about future movements in another.

Therefore:

SOURCE
   ↓
Risk / return shock
   ↓
TARGET
   ↓
Potential spillover

Assets with high outgoing connectivity are potential
RISK TRANSMITTERS.

Assets with high incoming connectivity are potential
RISK RECEIVERS.

The difference between the two produces a
NET CONTAGION SCORE.

This allows the market to be viewed as a
DIRECTED FINANCIAL NETWORK rather than a simple
correlation matrix.
""")

# ------------------------------------------------
# 15. COMPLETION
# ------------------------------------------------

print("\n" + "=" * 70)
print("PART 2 COMPLETE")
print("=" * 70)

print("""
Saved:

1. P2_All_Spillover_Tests.csv
2. P2_Significant_Spillovers.csv
3. P2_Contagion_Network_Ranking.csv

NEXT:
PART 3 → SYSTEMIC RISK & STRESS TESTING
""")
```

```python
# ================================================================
# PART 3: SYSTEMIC RISK & STRESS TESTING
# Financial Contagion Network
# ================================================================

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import networkx as nx

print("=" * 70)
print("PART 3: SYSTEMIC RISK & STRESS TESTING")
print("=" * 70)

# ------------------------------------------------
# 1. LOAD PART 1 & PART 2 RESULTS
# ------------------------------------------------

returns = pd.read_csv(
    "/content/P1_MultiAsset_Returns.csv",
    index_col=0,
    parse_dates=True
)

correlation = pd.read_csv(
    "/content/P1_Correlation_Matrix.csv",
    index_col=0
)

network_results = pd.read_csv(
    "/content/P2_Contagion_Network_Ranking.csv"
)

spillovers = pd.read_csv(
    "/content/P2_Significant_Spillovers.csv"
)

returns = returns.dropna()

assets = list(returns.columns)

print("\nAssets:", assets)

print(
    "Return observations:",
    len(returns)
)

print(
    "Significant spillover links:",
    len(spillovers)
)

# ------------------------------------------------
# 2. BUILD DIRECTED CONTAGION NETWORK
# ------------------------------------------------

G = nx.DiGraph()

for asset in assets:
    G.add_node(asset)

for _, row in spillovers.iterrows():

    source = row["source"]
    target = row["target"]
    p_value = row["min_p_value"]

    weight = -np.log10(
        max(p_value, 1e-10)
    )

    G.add_edge(
        source,
        target,
        weight=weight,
        p_value=p_value
    )

print("\n" + "=" * 70)
print("CONTAGION NETWORK")
print("=" * 70)

print("Nodes:", G.number_of_nodes())
print("Edges:", G.number_of_edges())

# ------------------------------------------------
# 3. SYSTEMIC IMPORTANCE
# ------------------------------------------------

out_strength = dict(
    G.out_degree(weight="weight")
)

in_strength = dict(
    G.in_degree(weight="weight")
)

pagerank = nx.pagerank(
    G,
    weight="weight"
)

betweenness = nx.betweenness_centrality(
    G,
    weight="weight"
)

systemic = pd.DataFrame({
    "asset": assets,

    "risk_transmission":
        [
            out_strength.get(
                x, 0
            )
            for x in assets
        ],

    "risk_absorption":
        [
            in_strength.get(
                x, 0
            )
            for x in assets
        ],

    "network_centrality":
        [
            pagerank.get(
                x, 0
            )
            for x in assets
        ],

    "bridge_centrality":
        [
            betweenness.get(
                x, 0
            )
            for x in assets
        ]
})

# ------------------------------------------------
# 4. NORMALIZE SYSTEMIC COMPONENTS
# ------------------------------------------------

def normalize(series):

    minimum = series.min()
    maximum = series.max()

    if maximum == minimum:
        return pd.Series(
            0.5,
            index=series.index
        )

    return (
        series - minimum
    ) / (
        maximum - minimum
    )


systemic[
    "transmission_norm"
] = normalize(
    systemic[
        "risk_transmission"
    ]
)

systemic[
    "centrality_norm"
] = normalize(
    systemic[
        "network_centrality"
    ]
)

systemic[
    "bridge_norm"
] = normalize(
    systemic[
        "bridge_centrality"
    ]
)

# ------------------------------------------------
# 5. VOLATILITY RISK
# ------------------------------------------------

annualized_volatility = (
    returns.std()
    * np.sqrt(365)
)

volatility_df = pd.DataFrame({
    "asset": annualized_volatility.index,
    "annualized_volatility":
        annualized_volatility.values
})

volatility_df[
    "volatility_norm"
] = normalize(
    volatility_df[
        "annualized_volatility"
    ]
)

systemic = systemic.merge(
    volatility_df,
    on="asset"
)

# ------------------------------------------------
# 6. FINAL SYSTEMIC RISK SCORE
# ------------------------------------------------

systemic["systemic_risk_score"] = (

    0.35 *
    systemic["transmission_norm"]

    +

    0.25 *
    systemic["centrality_norm"]

    +

    0.20 *
    systemic["bridge_norm"]

    +

    0.20 *
    systemic["volatility_norm"]

)

systemic = systemic.sort_values(
    "systemic_risk_score",
    ascending=False
)

# ------------------------------------------------
# 7. SYSTEMIC RISK REGIME
# ------------------------------------------------

def systemic_regime(x):

    if x >= 0.75:
        return "EXTREME"

    elif x >= 0.50:
        return "HIGH"

    elif x >= 0.25:
        return "MODERATE"

    else:
        return "LOW"


systemic[
    "systemic_risk_regime"
] = (
    systemic[
        "systemic_risk_score"
    ]
    .apply(systemic_regime)
)

print("\n" + "=" * 70)
print("SYSTEMIC RISK RANKING")
print("=" * 70)

display(
    systemic.round(4)
)

# ------------------------------------------------
# 8. SHOCK PROPAGATION ENGINE
# ------------------------------------------------

print("\n" + "=" * 70)
print("SHOCK PROPAGATION")
print("=" * 70)

# Use absolute correlation as the
# propagation coefficient.

corr_matrix = correlation.abs()

# ------------------------------------------------
# 9. PROPAGATION FUNCTION
# ------------------------------------------------

def propagate_shock(
    shocked_asset,
    shock_percent,
    max_steps=3,
    decay=0.50
):

    shock = pd.Series(
        0.0,
        index=assets
    )

    shock[
        shocked_asset
    ] = shock_percent

    total_effect = shock.copy()

    current = shock.copy()

    for step in range(
        max_steps
    ):

        next_shock = pd.Series(
            0.0,
            index=assets
        )

        for source in assets:

            if current[source] == 0:
                continue

            for target in assets:

                if source == target:
                    continue

                transmission = (
                    corr_matrix
                    .loc[
                        source,
                        target
                    ]
                )

                effect = (
                    current[source]
                    *
                    transmission
                    *
                    decay
                )

                next_shock[target] += effect

        total_effect += next_shock

        current = next_shock

    return total_effect


# ------------------------------------------------
# 10. STRESS SCENARIOS
# ------------------------------------------------

shock_levels = [
    -0.05,
    -0.10,
    -0.20,
    -0.30
]

stress_results = []

for asset in assets:

    for shock in shock_levels:

        propagation = (
            propagate_shock(
                asset,
                shock
            )
        )

        for target in assets:

            if target == asset:
                continue

            stress_results.append({

                "shocked_asset":
                    asset,

                "shock":
                    shock,

                "affected_asset":
                    target,

                "estimated_propagated_shock":
                    propagation[target]
            })


stress = pd.DataFrame(
    stress_results
)

# ------------------------------------------------
# 11. SYSTEM-WIDE IMPACT
# ------------------------------------------------

system_impact = (
    stress
    .groupby(
        [
            "shocked_asset",
            "shock"
        ]
    )
    .agg(
        average_propagated_impact=(
            "estimated_propagated_shock",
            "mean"
        ),

        maximum_propagated_impact=(
            "estimated_propagated_shock",
            "max"
        ),

        total_absolute_impact=(
            "estimated_propagated_shock",
            lambda x:
                np.abs(x).sum()
        )
    )
    .reset_index()
)

print("\n" + "=" * 70)
print("SYSTEM-WIDE STRESS IMPACT")
print("=" * 70)

display(
    system_impact
    .sort_values(
        "total_absolute_impact",
        ascending=False
    )
    .head(20)
)

# ------------------------------------------------
# 12. WORST SHOCK ORIGINS
# ------------------------------------------------

worst_origins = (
    system_impact
    .groupby("shocked_asset")
    ["total_absolute_impact"]
    .mean()
    .sort_values(
        ascending=False
    )
    .reset_index()
)

worst_origins.columns = [
    "asset",
    "average_system_impact"
]

print("\n" + "=" * 70)
print("MOST SYSTEMIC SHOCK ORIGINS")
print("=" * 70)

display(
    worst_origins
)

# ------------------------------------------------
# 13. MOST VULNERABLE RECEIVERS
# ------------------------------------------------

vulnerable = (
    stress
    .groupby("affected_asset")
    ["estimated_propagated_shock"]
    .mean()
    .abs()
    .sort_values(
        ascending=False
    )
    .reset_index()
)

vulnerable.columns = [
    "asset",
    "average_received_shock"
]

print("\n" + "=" * 70)
print("MOST VULNERABLE ASSETS")
print("=" * 70)

display(
    vulnerable
)

# ------------------------------------------------
# 14. FINAL SYSTEMIC SCORE
# ------------------------------------------------

impact_score = (
    worst_origins
    .set_index("asset")
    ["average_system_impact"]
)

systemic = systemic.set_index(
    "asset"
)

systemic[
    "stress_impact"
] = systemic.index.map(
    impact_score
).fillna(0)

systemic[
    "stress_impact_norm"
] = normalize(
    systemic[
        "stress_impact"
    ]
)

# Add stress impact to final score
systemic[
    "final_systemic_risk"
] = (

    0.70 *
    systemic[
        "systemic_risk_score"
    ]

    +

    0.30 *
    systemic[
        "stress_impact_norm"
    ]
)

systemic[
    "final_systemic_regime"
] = (
    systemic[
        "final_systemic_risk"
    ]
    .apply(systemic_regime)
)

systemic = systemic.sort_values(
    "final_systemic_risk",
    ascending=False
)

print("\n" + "=" * 70)
print("FINAL SYSTEMIC RISK MODEL")
print("=" * 70)

display(
    systemic.round(4)
)

# ------------------------------------------------
# 15. TOP SYSTEMICALLY IMPORTANT ASSETS
# ------------------------------------------------

print("\n" + "=" * 70)
print("TOP SYSTEMICALLY IMPORTANT ASSETS")
print("=" * 70)

display(
    systemic[
        [
            "final_systemic_risk",
            "final_systemic_regime",
            "risk_transmission",
            "network_centrality",
            "bridge_centrality",
            "annualized_volatility"
        ]
    ].head(5)
)

# ------------------------------------------------
# 16. SAVE RESULTS
# ------------------------------------------------

systemic.to_csv(
    "/content/P3_Systemic_Risk_Ranking.csv"
)

stress.to_csv(
    "/content/P3_Stress_Test_Results.csv",
    index=False
)

system_impact.to_csv(
    "/content/P3_System_Wide_Stress_Impact.csv",
    index=False
)

worst_origins.to_csv(
    "/content/P3_Systemic_Shock_Origins.csv",
    index=False
)

vulnerable.to_csv(
    "/content/P3_Vulnerable_Assets.csv",
    index=False
)

# ------------------------------------------------
# 17. INTERPRETATION
# ------------------------------------------------

print("\n" + "=" * 70)
print("PART 3 INTERPRETATION")
print("=" * 70)

print("""
The systemic-risk model combines:

1. Risk transmission
2. Network centrality
3. Bridge/connector importance
4. Asset volatility
5. Stress-test impact

An asset can therefore be important for different reasons.

HIGH TRANSMITTER
→ Sends shocks to many other assets.

HIGH RECEIVER
→ Is vulnerable to external shocks.

HIGH CENTRALITY
→ Occupies an important position in the network.

HIGH BRIDGE CENTRALITY
→ Connects otherwise separate parts of the market.

HIGH STRESS IMPACT
→ A shock originating from this asset produces
  relatively large network-wide effects.

The final systemic-risk score combines these dimensions
to identify assets that could have disproportionate
importance during market stress.
""")

# ------------------------------------------------
# 18. COMPLETION
# ------------------------------------------------

print("\n" + "=" * 70)
print("PART 3 COMPLETE")
print("=" * 70)

print("""
Saved:

1. P3_Systemic_Risk_Ranking.csv
2. P3_Stress_Test_Results.csv
3. P3_System_Wide_Stress_Impact.csv
4. P3_Systemic_Shock_Origins.csv
5. P3_Vulnerable_Assets.csv

NEXT:
PART 4 → DYNAMIC CONTAGION DASHBOARD & FINAL DECISION FRAMEWORK
""")
```

```python
# ================================================================
# PART 4: DYNAMIC CONTAGION DASHBOARD
# FINAL SYSTEMIC-RISK DECISION FRAMEWORK
# ================================================================

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import networkx as nx

print("=" * 70)
print("PART 4: DYNAMIC CONTAGION DASHBOARD")
print("=" * 70)

# ------------------------------------------------
# 1. LOAD PREVIOUS RESULTS
# ------------------------------------------------

returns = pd.read_csv(
    "/content/P1_MultiAsset_Returns.csv",
    index_col=0,
    parse_dates=True
).dropna()

systemic = pd.read_csv(
    "/content/P3_Systemic_Risk_Ranking.csv"
)

stress = pd.read_csv(
    "/content/P3_Stress_Test_Results.csv"
)

impact = pd.read_csv(
    "/content/P3_System_Wide_Stress_Impact.csv"
)

print("\nReturns:", returns.shape)
print("Systemic ranking:", systemic.shape)
print("Stress observations:", stress.shape)

# ------------------------------------------------
# 2. MARKET-WIDE RISK INDICATORS
# ------------------------------------------------

portfolio_returns = returns.mean(axis=1)

rolling_volatility = (
    portfolio_returns
    .rolling(30)
    .std()
)

rolling_drawdown = (
    portfolio_returns
    .rolling(30)
    .apply(
        lambda x:
        (x[-1] / np.max(x)) - 1
        if np.max(x) != 0
        else 0
    )
)

market_risk = pd.DataFrame({
    "portfolio_return": portfolio_returns,
    "rolling_volatility": rolling_volatility,
    "rolling_drawdown": rolling_drawdown
}).dropna()

# ------------------------------------------------
# 3. DYNAMIC MARKET REGIME
# ------------------------------------------------

vol_threshold = (
    market_risk[
        "rolling_volatility"
    ].quantile(0.75)
)

drawdown_threshold = (
    market_risk[
        "rolling_drawdown"
    ].quantile(0.25)
)

def market_regime(row):

    if (
        row["rolling_volatility"]
        >= vol_threshold
        and
        row["rolling_drawdown"]
        <= drawdown_threshold
    ):
        return "STRESSED"

    elif (
        row["rolling_volatility"]
        >= vol_threshold
    ):
        return "HIGH VOLATILITY"

    else:
        return "NORMAL"


market_risk[
    "market_regime"
] = market_risk.apply(
    market_regime,
    axis=1
)

print("\n" + "=" * 70)
print("MARKET REGIME DISTRIBUTION")
print("=" * 70)

display(
    market_risk[
        "market_regime"
    ]
    .value_counts()
)

# ------------------------------------------------
# 4. SYSTEMIC RISK SUMMARY
# ------------------------------------------------

systemic["final_systemic_risk"] = pd.to_numeric(
    systemic["final_systemic_risk"],
    errors="coerce"
)

systemic = systemic.dropna(
    subset=["final_systemic_risk"]
)

systemic = systemic.sort_values(
    "final_systemic_risk",
    ascending=False
)

print("\n" + "=" * 70)
print("TOP SYSTEMICALLY IMPORTANT ASSETS")
print("=" * 70)

display(
    systemic[
        [
            "asset",
            "final_systemic_risk",
            "final_systemic_regime",
            "risk_transmission",
            "risk_absorption",
            "network_centrality",
            "bridge_centrality",
            "stress_impact"
        ]
    ].head(10).round(4)
)

# ------------------------------------------------
# 5. TRANSMITTER VS RECEIVER MAP
# ------------------------------------------------

print("\n" + "=" * 70)
print("RISK TRANSMITTER VS RECEIVER")
print("=" * 70)

plt.figure(figsize=(10, 7))

plt.scatter(
    systemic[
        "risk_transmission"
    ],
    systemic[
        "risk_absorption"
    ],
    s=100
)

for _, row in systemic.iterrows():

    plt.annotate(
        row["asset"],
        (
            row["risk_transmission"],
            row["risk_absorption"]
        )
    )

plt.xlabel(
    "Risk Transmission Strength"
)

plt.ylabel(
    "Risk Absorption Strength"
)

plt.title(
    "Financial Contagion: Transmitters vs Receivers"
)

plt.grid(alpha=0.3)

plt.show()

# ------------------------------------------------
# 6. SYSTEMIC RISK RANKING
# ------------------------------------------------

top_n = min(
    10,
    len(systemic)
)

top_systemic = systemic.head(
    top_n
).sort_values(
    "final_systemic_risk"
)

plt.figure(figsize=(10, 6))

plt.barh(
    top_systemic["asset"],
    top_systemic["final_systemic_risk"]
)

plt.xlabel(
    "Final Systemic Risk Score"
)

plt.ylabel(
    "Asset"
)

plt.title(
    "Top Systemically Important Assets"
)

plt.grid(
    axis="x",
    alpha=0.3
)

plt.show()

# ------------------------------------------------
# 7. STRESS IMPACT RANKING
# ------------------------------------------------

stress_rank = (
    impact
    .groupby("shocked_asset")
    [
        "total_absolute_impact"
    ]
    .mean()
    .sort_values(
        ascending=False
    )
)

plt.figure(figsize=(10, 6))

stress_rank.head(10).sort_values().plot(
    kind="barh"
)

plt.xlabel(
    "Average Network-Wide Stress Impact"
)

plt.ylabel(
    "Shock Origin"
)

plt.title(
    "Assets With Highest System-Wide Stress Impact"
)

plt.grid(
    axis="x",
    alpha=0.3
)

plt.show()

# ------------------------------------------------
# 8. FINAL DECISION ENGINE
# ------------------------------------------------

def decision(row):

    risk = row[
        "final_systemic_risk"
    ]

    transmission = row[
        "risk_transmission"
    ]

    stress = row[
        "stress_impact"
    ]

    if risk >= 0.75:

        return (
            "CRITICAL: REDUCE EXPOSURE "
            "AND MONITOR CONTAGION"
        )

    elif risk >= 0.50:

        return (
            "HIGH RISK: TIGHTEN RISK LIMITS "
            "AND REDUCE CONCENTRATION"
        )

    elif (
        risk >= 0.25
    ):

        return (
            "MODERATE: MONITOR "
            "SPILLOVER CONDITIONS"
        )

    else:

        return (
            "LOW: NORMAL RISK MONITORING"
        )


systemic[
    "strategic_decision"
] = systemic.apply(
    decision,
    axis=1
)

# ------------------------------------------------
# 9. FINAL DECISION TABLE
# ------------------------------------------------

print("\n" + "=" * 70)
print("FINAL STRATEGIC DECISION")
print("=" * 70)

display(
    systemic[
        [
            "asset",
            "final_systemic_risk",
            "final_systemic_regime",
            "strategic_decision"
        ]
    ].round(4)
)

# ------------------------------------------------
# 10. PORTFOLIO-LEVEL SYSTEMIC RISK
# ------------------------------------------------

portfolio_systemic_risk = (
    systemic[
        "final_systemic_risk"
    ].mean()
)

maximum_systemic_risk = (
    systemic[
        "final_systemic_risk"
    ].max()
)

high_risk_count = (
    systemic[
        "final_systemic_risk"
    ] >= 0.50
).sum()

extreme_risk_count = (
    systemic[
        "final_systemic_risk"
    ] >= 0.75
).sum()

print("\n" + "=" * 70)
print("PORTFOLIO SYSTEMIC RISK SUMMARY")
print("=" * 70)

print(
    "Average systemic risk:",
    round(
        portfolio_systemic_risk,
        4
    )
)

print(
    "Maximum systemic risk:",
    round(
        maximum_systemic_risk,
        4
    )
)

print(
    "High-risk assets:",
    high_risk_count
)

print(
    "Extreme-risk assets:",
    extreme_risk_count
)

# ------------------------------------------------
# 11. FINAL FRAMEWORK
# ------------------------------------------------

print("\n" + "=" * 70)
print("FINAL CONTAGION FRAMEWORK")
print("=" * 70)

print("""
MARKET RETURNS
      ↓
CORRELATION STRUCTURE
      ↓
DIRECTIONAL SPILLOVER
      ↓
CONTAGION NETWORK
      ↓
RISK TRANSMITTERS / RECEIVERS
      ↓
SYSTEMIC IMPORTANCE
      ↓
STRESS TESTING
      ↓
NETWORK-WIDE IMPACT
      ↓
FINAL SYSTEMIC RISK
      ↓
STRATEGIC DECISION
""")

# ------------------------------------------------
# 12. SAVE FINAL OUTPUT
# ------------------------------------------------

systemic.to_csv(
    "/content/P4_Final_Systemic_Risk_Dashboard.csv",
    index=False
)

market_risk.to_csv(
    "/content/P4_Dynamic_Market_Regimes.csv"
)

stress_rank.to_csv(
    "/content/P4_Stress_Origin_Ranking.csv"
)

# ------------------------------------------------
# 13. FINAL PROJECT STATISTICS
# ------------------------------------------------

print("\n" + "=" * 70)
print("FINAL PROJECT STATISTICS")
print("=" * 70)

print(
    "Assets analysed:",
    len(returns.columns)
)

print(
    "Return observations:",
    len(returns)
)

print(
    "Systemic assets ranked:",
    len(systemic)
)

print(
    "Stress-test observations:",
    len(stress)
)

print(
    "Average systemic risk:",
    round(
        portfolio_systemic_risk,
        4
    )
)

print(
    "Highest systemic risk:",
    round(
        maximum_systemic_risk,
        4
    )
)

print("\n" + "=" * 70)
print("PROJECT COMPLETE")
print("=" * 70)

print("""
Saved:

1. P4_Final_Systemic_Risk_Dashboard.csv
2. P4_Dynamic_Market_Regimes.csv
3. P4_Stress_Origin_Ranking.csv

FINAL PROJECT:

MULTI-ASSET CONTAGION &
SYSTEMIC RISK ANALYSIS

Framework:

Returns
   ↓
Correlation
   ↓
Spillover
   ↓
Contagion Network
   ↓
Systemic Risk
   ↓
Stress Testing
   ↓
Strategic Decision
""")
```
