```python
# ============================================================
# QUALITY-ADJUSTED GROWTH FACTOR ALGORITHMIC TRADING
# ============================================================

!pip -q install yfinance

import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings

warnings.filterwarnings("ignore")

# ============================================================
# 1. CONFIGURATION
# ============================================================

START_DATE = "2015-01-01"
INITIAL_CAPITAL = 1_000_000

STOCKS = [
    "RELIANCE.NS",
    "HDFCBANK.NS",
    "ICICIBANK.NS",
    "INFY.NS",
    "TCS.NS",
    "ITC.NS",
    "BHARTIARTL.NS",
    "SBIN.NS",
    "LT.NS",
    "AXISBANK.NS",
    "KOTAKBANK.NS",
    "HINDUNILVR.NS",
    "MARUTI.NS",
    "SUNPHARMA.NS",
    "M&M.NS",
    "TATAMOTORS.NS",
    "ADANIENT.NS",
    "NTPC.NS",
    "POWERGRID.NS",
    "TATASTEEL.NS"
]

BENCHMARK = "^NSEI"

N_LONG = 5
N_SHORT = 5


# ============================================================
# 2. COLLECT FUNDAMENTAL DATA USING TICKER.INFO
# ============================================================

fundamentals = []

print("Collecting fundamental information...\n")

for ticker in STOCKS:

    try:

        stock = yf.Ticker(ticker)
        info = stock.info

        trailing_eps = info.get("trailingEps")
        forward_eps = info.get("forwardEps")

        # EPS Growth Proxy
        if (
            trailing_eps is not None
            and forward_eps is not None
            and trailing_eps != 0
        ):
            eps_growth = (
                forward_eps / trailing_eps
            ) - 1
        else:
            eps_growth = np.nan

        fundamentals.append({

            "Ticker": ticker,

            "Company":
                info.get("longName", ticker),

            "Sector":
                info.get("sector", "Unknown"),

            "Industry":
                info.get("industry", "Unknown"),

            "MarketCap":
                info.get("marketCap", np.nan),

            # Growth
            "RevenueGrowth":
                info.get("revenueGrowth", np.nan),

            "EarningsGrowth":
                info.get("earningsGrowth", np.nan),

            "EPS_Growth":
                eps_growth,

            # Quality
            "ROE":
                info.get("returnOnEquity", np.nan),

            "ROA":
                info.get("returnOnAssets", np.nan),

            "OperatingMargin":
                info.get("operatingMargins", np.nan),

            "ProfitMargin":
                info.get("profitMargins", np.nan),

            "DebtToEquity":
                info.get("debtToEquity", np.nan),

            "CurrentRatio":
                info.get("currentRatio", np.nan),

            "Beta":
                info.get("beta", np.nan),

            "PE":
                info.get("trailingPE", np.nan),

            "TrailingEPS":
                trailing_eps,

            "ForwardEPS":
                forward_eps
        })

        print(f"✓ {ticker}")

    except Exception as e:

        print(f"✗ {ticker}: {e}")


fundamentals = pd.DataFrame(fundamentals)

print("\nFundamental dataset:")
display(fundamentals)


# ============================================================
# 3. CLEAN DATA
# ============================================================

factor_columns = [
    "RevenueGrowth",
    "EarningsGrowth",
    "EPS_Growth",
    "ROE",
    "ROA",
    "OperatingMargin",
    "ProfitMargin"
]

for col in factor_columns:

    fundamentals[col] = pd.to_numeric(
        fundamentals[col],
        errors="coerce"
    )

    # Convert decimal values to percentages
    fundamentals[col] *= 100


# ============================================================
# 4. WINSORIZE EXTREME VALUES
# ============================================================

for col in factor_columns:

    if fundamentals[col].notna().sum() >= 5:

        lower = fundamentals[col].quantile(0.05)
        upper = fundamentals[col].quantile(0.95)

        fundamentals[col] = fundamentals[col].clip(
            lower,
            upper
        )


# ============================================================
# 5. PERCENTILE RANKING FUNCTION
# ============================================================

def percentile_score(series):

    return series.rank(
        pct=True,
        method="average"
    )


# ============================================================
# 6. GROWTH FACTOR SCORES
# ============================================================

fundamentals["RevenueGrowth_Score"] = (
    percentile_score(
        fundamentals["RevenueGrowth"]
    )
)

fundamentals["EarningsGrowth_Score"] = (
    percentile_score(
        fundamentals["EarningsGrowth"]
    )
)

fundamentals["EPSGrowth_Score"] = (
    percentile_score(
        fundamentals["EPS_Growth"]
    )
)


# ============================================================
# 7. QUALITY FACTOR SCORES
# ============================================================

fundamentals["ROE_Score"] = (
    percentile_score(
        fundamentals["ROE"]
    )
)

fundamentals["ROA_Score"] = (
    percentile_score(
        fundamentals["ROA"]
    )
)

fundamentals["OperatingMargin_Score"] = (
    percentile_score(
        fundamentals["OperatingMargin"]
    )
)

fundamentals["ProfitMargin_Score"] = (
    percentile_score(
        fundamentals["ProfitMargin"]
    )
)


# ============================================================
# 8. COMPOSITE GROWTH SCORE
# ============================================================

fundamentals["Growth_Score"] = (

    0.40 * fundamentals["RevenueGrowth_Score"]

    + 0.35 * fundamentals["EarningsGrowth_Score"]

    + 0.25 * fundamentals["EPSGrowth_Score"]
)


# ============================================================
# 9. COMPOSITE QUALITY SCORE
# ============================================================

fundamentals["Quality_Score"] = (

    0.30 * fundamentals["ROE_Score"]

    + 0.20 * fundamentals["ROA_Score"]

    + 0.25 * fundamentals["OperatingMargin_Score"]

    + 0.25 * fundamentals["ProfitMargin_Score"]
)


# ============================================================
# 10. QUALITY-ADJUSTED GROWTH SCORE
# ============================================================

fundamentals["QAG_Score"] = (

    0.60 * fundamentals["Growth_Score"]

    + 0.40 * fundamentals["Quality_Score"]
)


# ============================================================
# 11. FINAL CROSS-SECTIONAL RANKING
# ============================================================

fundamentals["QAG_Rank"] = (
    fundamentals["QAG_Score"]
    .rank(
        ascending=False,
        method="first"
    )
)

ranking = fundamentals.sort_values(
    "QAG_Rank"
).reset_index(drop=True)


print("\n==============================")
print("QUALITY-ADJUSTED GROWTH RANKING")
print("==============================")

display(
    ranking[
        [
            "Ticker",
            "Company",
            "Sector",
            "Growth_Score",
            "Quality_Score",
            "QAG_Score",
            "QAG_Rank"
        ]
    ]
)


# ============================================================
# 12. LONG / SHORT PORTFOLIO
# ============================================================

valid_ranking = ranking.dropna(
    subset=["QAG_Score"]
)

long_stocks = (
    valid_ranking
    .head(N_LONG)["Ticker"]
    .tolist()
)

short_stocks = (
    valid_ranking
    .tail(N_SHORT)["Ticker"]
    .tolist()
)

print("\nLONG PORTFOLIO")
print("----------------")

for x in long_stocks:
    print(x)

print("\nSHORT PORTFOLIO")
print("----------------")

for x in short_stocks:
    print(x)


# ============================================================
# 13. DOWNLOAD HISTORICAL PRICE DATA
# ============================================================

print("\nDownloading historical prices...")

price_data = yf.download(
    STOCKS + [BENCHMARK],
    start=START_DATE,
    auto_adjust=True,
    progress=False
)

if isinstance(price_data.columns, pd.MultiIndex):

    close = price_data["Close"]

else:

    close = price_data[["Close"]]

close = close.dropna(
    how="all"
)

print("Price data shape:", close.shape)


# ============================================================
# 14. MONTHLY RETURNS
# ============================================================

monthly_prices = (
    close
    .resample("ME")
    .last()
)

monthly_returns = (
    monthly_prices
    .pct_change()
    .dropna(how="all")
)


# ============================================================
# 15. LONG / SHORT RETURNS
# ============================================================

available_long = [
    x for x in long_stocks
    if x in monthly_returns.columns
]

available_short = [
    x for x in short_stocks
    if x in monthly_returns.columns
]

long_returns = (
    monthly_returns[available_long]
    .mean(axis=1)
)

short_returns = (
    monthly_returns[available_short]
    .mean(axis=1)
)

# Long-short
strategy_returns = (
    long_returns - short_returns
).dropna()

# Long-only
long_only_returns = (
    long_returns
).dropna()

# Benchmark
benchmark_returns = (
    monthly_returns[BENCHMARK]
).dropna()


# ============================================================
# 16. ALIGN RETURNS
# ============================================================

common_index = (
    strategy_returns.index
    .intersection(benchmark_returns.index)
)

strategy_returns = strategy_returns.loc[
    common_index
]

long_only_returns = long_only_returns.loc[
    long_only_returns.index.intersection(common_index)
]

benchmark_returns = benchmark_returns.loc[
    common_index
]


# ============================================================
# 17. EQUITY CURVES
# ============================================================

strategy_equity = (
    1 + strategy_returns
).cumprod()

long_only_equity = (
    1 + long_only_returns
).cumprod()

benchmark_equity = (
    1 + benchmark_returns
).cumprod()


# ============================================================
# 18. PERFORMANCE METRICS
# ============================================================

def performance_metrics(returns):

    returns = returns.dropna()

    if len(returns) < 2:
        return {}

    cumulative_return = (
        1 + returns
    ).prod() - 1

    annual_return = (
        (1 + cumulative_return)
        ** (12 / len(returns))
        - 1
    )

    annual_volatility = (
        returns.std() * np.sqrt(12)
    )

    sharpe = (
        annual_return / annual_volatility
        if annual_volatility > 0
        else np.nan
    )

    downside_returns = returns[
        returns < 0
    ]

    downside_volatility = (
        downside_returns.std()
        * np.sqrt(12)
        if len(downside_returns) > 1
        else np.nan
    )

    sortino = (
        annual_return / downside_volatility
        if downside_volatility > 0
        else np.nan
    )

    equity = (
        1 + returns
    ).cumprod()

    drawdown = (
        equity /
        equity.cummax()
    ) - 1

    max_drawdown = drawdown.min()

    win_rate = (
        returns > 0
    ).mean()

    return {

        "Cumulative Return":
            cumulative_return,

        "Annual Return":
            annual_return,

        "Annual Volatility":
            annual_volatility,

        "Sharpe Ratio":
            sharpe,

        "Sortino Ratio":
            sortino,

        "Maximum Drawdown":
            max_drawdown,

        "Win Rate":
            win_rate,

        "Observations":
            len(returns)
    }


# ============================================================
# 19. PERFORMANCE COMPARISON
# ============================================================

performance = pd.DataFrame({

    "QAG Long-Short":
        performance_metrics(
            strategy_returns
        ),

    "QAG Long-Only":
        performance_metrics(
            long_only_returns
        ),

    "NIFTY 50":
        performance_metrics(
            benchmark_returns
        )

}).T

print("\n==============================")
print("PERFORMANCE COMPARISON")
print("==============================")

display(
    performance.style.format({

        "Cumulative Return": "{:.2%}",

        "Annual Return": "{:.2%}",

        "Annual Volatility": "{:.2%}",

        "Sharpe Ratio": "{:.2f}",

        "Sortino Ratio": "{:.2f}",

        "Maximum Drawdown": "{:.2%}",

        "Win Rate": "{:.2%}"
    })
)


# ============================================================
# 20. EQUITY CURVE
# ============================================================

plt.figure(figsize=(14, 7))

plt.plot(
    strategy_equity,
    label="QAG Long-Short"
)

plt.plot(
    long_only_equity,
    label="QAG Long-Only"
)

plt.plot(
    benchmark_equity,
    label="NIFTY 50"
)

plt.title(
    "Quality-Adjusted Growth Factor Strategy"
)

plt.xlabel("Date")
plt.ylabel("Growth of ₹1")

plt.legend()
plt.grid(alpha=0.3)

plt.show()


# ============================================================
# 21. DRAWDOWN ANALYSIS
# ============================================================

strategy_dd = (
    strategy_equity /
    strategy_equity.cummax()
) - 1

long_only_dd = (
    long_only_equity /
    long_only_equity.cummax()
) - 1

benchmark_dd = (
    benchmark_equity /
    benchmark_equity.cummax()
) - 1


plt.figure(figsize=(14, 6))

plt.plot(
    strategy_dd,
    label="QAG Long-Short"
)

plt.plot(
    long_only_dd,
    label="QAG Long-Only"
)

plt.plot(
    benchmark_dd,
    label="NIFTY 50"
)

plt.title(
    "Drawdown Comparison"
)

plt.xlabel("Date")
plt.ylabel("Drawdown")

plt.legend()
plt.grid(alpha=0.3)

plt.show()


# ============================================================
# 22. ROLLING 12-MONTH SHARPE
# ============================================================

rolling_mean = (
    strategy_returns
    .rolling(12)
    .mean()
)

rolling_std = (
    strategy_returns
    .rolling(12)
    .std()
)

rolling_sharpe = (
    rolling_mean /
    rolling_std
) * np.sqrt(12)


plt.figure(figsize=(14, 6))

plt.plot(
    rolling_sharpe,
    label="12-Month Rolling Sharpe"
)

plt.axhline(
    0,
    linestyle="--"
)

plt.title(
    "QAG Strategy Rolling Sharpe Ratio"
)

plt.xlabel("Date")
plt.ylabel("Sharpe Ratio")

plt.legend()
plt.grid(alpha=0.3)

plt.show()


# ============================================================
# 23. GROWTH VS QUALITY
# ============================================================

plt.figure(figsize=(10, 7))

plt.scatter(
    fundamentals["Growth_Score"],
    fundamentals["Quality_Score"],
    s=100
)

for _, row in fundamentals.iterrows():

    if pd.notna(row["Growth_Score"]) and \
       pd.notna(row["Quality_Score"]):

        plt.annotate(
            row["Ticker"].replace(".NS", ""),
            (
                row["Growth_Score"],
                row["Quality_Score"]
            ),
            xytext=(5, 5),
            textcoords="offset points"
        )

plt.xlabel("Growth Score")
plt.ylabel("Quality Score")

plt.title(
    "Growth vs Quality Factor Map"
)

plt.grid(alpha=0.3)

plt.show()


# ============================================================
# 24. FACTOR SCORE DISTRIBUTION
# ============================================================

plt.figure(figsize=(12, 6))

plt.hist(
    fundamentals["QAG_Score"].dropna(),
    bins=10
)

plt.xlabel(
    "Quality-Adjusted Growth Score"
)

plt.ylabel(
    "Number of Stocks"
)

plt.title(
    "QAG Score Distribution"
)

plt.grid(alpha=0.3)

plt.show()


# ============================================================
# 25. SECTOR EXPOSURE
# ============================================================

long_sector = (
    fundamentals[
        fundamentals["Ticker"].isin(
            long_stocks
        )
    ]["Sector"]
    .value_counts()
)

short_sector = (
    fundamentals[
        fundamentals["Ticker"].isin(
            short_stocks
        )
    ]["Sector"]
    .value_counts()
)

sector_table = pd.concat(
    [
        long_sector.rename("Long"),
        short_sector.rename("Short")
    ],
    axis=1
).fillna(0)


print("\nSector Exposure")

display(sector_table)


sector_table.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title(
    "QAG Portfolio Sector Exposure"
)

plt.ylabel(
    "Number of Stocks"
)

plt.xlabel(
    "Sector"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.show()


# ============================================================
# 26. ANNUAL RETURNS
# ============================================================

annual_returns = pd.DataFrame({

    "QAG_Long_Short":
        strategy_returns,

    "QAG_Long_Only":
        long_only_returns,

    "NIFTY_50":
        benchmark_returns

}).groupby(
    lambda x: x.year
).apply(
    lambda x: (1 + x).prod() - 1
)

print("\nAnnual Returns")

display(
    annual_returns.style.format(
        "{:.2%}"
    )
)


# ============================================================
# 27. FACTOR CORRELATION
# ============================================================

factor_corr = fundamentals[
    [
        "RevenueGrowth",
        "EarningsGrowth",
        "EPS_Growth",
        "ROE",
        "ROA",
        "OperatingMargin",
        "ProfitMargin",
        "Growth_Score",
        "Quality_Score",
        "QAG_Score"
    ]
].corr()


plt.figure(figsize=(12, 9))

plt.imshow(
    factor_corr,
    aspect="auto"
)

plt.colorbar(
    label="Correlation"
)

plt.xticks(
    range(len(factor_corr.columns)),
    factor_corr.columns,
    rotation=90
)

plt.yticks(
    range(len(factor_corr.columns)),
    factor_corr.columns
)

plt.title(
    "QAG Factor Correlation Matrix"
)

plt.tight_layout()

plt.show()


# ============================================================
# 28. FINAL STOCK RANKING
# ============================================================

final_ranking = fundamentals[
    [
        "Ticker",
        "Company",
        "Sector",
        "RevenueGrowth",
        "EarningsGrowth",
        "EPS_Growth",
        "ROE",
        "ROA",
        "OperatingMargin",
        "ProfitMargin",
        "Growth_Score",
        "Quality_Score",
        "QAG_Score",
        "QAG_Rank"
    ]
].sort_values(
    "QAG_Rank"
)


print("\n==============================")
print("FINAL QUALITY-ADJUSTED GROWTH RANKING")
print("==============================")

display(final_ranking)


# ============================================================
# 29. EXPORT RESULTS
# ============================================================

fundamentals.to_csv(
    "Quality_Adjusted_Growth_Factor_Ranking.csv",
    index=False
)

performance.to_csv(
    "Quality_Adjusted_Growth_Performance.csv"
)

annual_returns.to_csv(
    "Quality_Adjusted_Growth_Annual_Returns.csv"
)

sector_table.to_csv(
    "Quality_Adjusted_Growth_Sector_Exposure.csv"
)

strategy_returns.to_csv(
    "Quality_Adjusted_Growth_Long_Short_Returns.csv"
)

long_only_returns.to_csv(
    "Quality_Adjusted_Growth_Long_Only_Returns.csv"
)

print("\n========================================")
print("RESEARCH PIPELINE COMPLETED SUCCESSFULLY")
print("========================================")

print("\nFiles exported:")
print("✓ Quality_Adjusted_Growth_Factor_Ranking.csv")
print("✓ Quality_Adjusted_Growth_Performance.csv")
print("✓ Quality_Adjusted_Growth_Annual_Returns.csv")
print("✓ Quality_Adjusted_Growth_Sector_Exposure.csv")
print("✓ Quality_Adjusted_Growth_Long_Short_Returns.csv")
print("✓ Quality_Adjusted_Growth_Long_Only_Returns.csv")
```
