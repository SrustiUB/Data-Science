```python
# ============================================================
# VALUE FACTOR ALGORITHMIC TRADING
# ============================================================

!pip -q install yfinance

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import yfinance as yf


# ============================================================
# 1. STOCK UNIVERSE
# ============================================================

TICKERS = [
    "RELIANCE.NS",
    "HDFCBANK.NS",
    "ICICIBANK.NS",
    "INFY.NS",
    "TCS.NS",
    "ITC.NS",
    "BHARTIARTL.NS",
    "SBIN.NS",
    "LT.NS",
    "AXISBANK.NS",
    "KOTAKBANK.NS",
    "HINDUNILVR.NS",
    "MARUTI.NS",
    "SUNPHARMA.NS",
    "M&M.NS",
    "TATAMOTORS.NS",
    "ADANIENT.NS",
    "NTPC.NS",
    "POWERGRID.NS",
    "TATASTEEL.NS"
]

START_DATE = "2015-01-01"

print("Number of stocks:", len(TICKERS))


# ============================================================
# 2. GET FUNDAMENTAL INFORMATION
# ============================================================

fundamentals = []

print("\nCollecting fundamental information...\n")

for ticker in TICKERS:

    try:

        info = yf.Ticker(ticker).info

        fundamentals.append({

            "Ticker": ticker,

            "Company":
                info.get("longName", ticker),

            "Sector":
                info.get("sector", "N/A"),

            "Industry":
                info.get("industry", "N/A"),

            "MarketCap":
                info.get("marketCap", np.nan),

            "PE":
                info.get("trailingPE", np.nan),

            "ForwardPE":
                info.get("forwardPE", np.nan),

            "PB":
                info.get("priceToBook", np.nan),

            "EV_EBITDA":
                info.get("enterpriseToEbitda", np.nan),

            "EarningsYield":
                (
                    1 / info.get("trailingPE")
                    if info.get("trailingPE")
                    and info.get("trailingPE") > 0
                    else np.nan
                ),

            "FCF":
                info.get("freeCashflow", np.nan),

            "TotalRevenue":
                info.get("totalRevenue", np.nan),

            "EnterpriseValue":
                info.get(
                    "enterpriseValue",
                    np.nan
                )
        })

    except Exception:

        fundamentals.append({

            "Ticker": ticker,
            "Company": ticker,
            "Sector": "N/A",
            "Industry": "N/A",
            "MarketCap": np.nan,
            "PE": np.nan,
            "ForwardPE": np.nan,
            "PB": np.nan,
            "EV_EBITDA": np.nan,
            "EarningsYield": np.nan,
            "FCF": np.nan,
            "TotalRevenue": np.nan,
            "EnterpriseValue": np.nan
        })


fundamental_df = pd.DataFrame(
    fundamentals
)

# ------------------------------------------------------------
# FCF Yield
# ------------------------------------------------------------

fundamental_df["FCF_Yield"] = (
    fundamental_df["FCF"]
    /
    fundamental_df["MarketCap"]
)

# Clean extreme values

for column in [
    "PE",
    "ForwardPE",
    "PB",
    "EV_EBITDA",
    "FCF_Yield",
    "EarningsYield"
]:

    fundamental_df[column] = (
        pd.to_numeric(
            fundamental_df[column],
            errors="coerce"
        )
    )

print("\nFundamental Dataset:")
display(fundamental_df)


# ============================================================
# 3. FILTER INVALID FUNDAMENTALS
# ============================================================

value_df = fundamental_df.copy()

# Positive valuation multiples
value_df.loc[
    value_df["PE"] <= 0,
    "PE"
] = np.nan

value_df.loc[
    value_df["PB"] <= 0,
    "PB"
] = np.nan

value_df.loc[
    value_df["EV_EBITDA"] <= 0,
    "EV_EBITDA"
] = np.nan

# Remove extreme observations

for column in [
    "PE",
    "PB",
    "EV_EBITDA",
    "FCF_Yield",
    "EarningsYield"
]:

    lower = value_df[column].quantile(
        0.02
    )

    upper = value_df[column].quantile(
        0.98
    )

    value_df[column] = (
        value_df[column]
        .clip(lower, upper)
    )


# ============================================================
# 4. FACTOR RANKING
# ============================================================

# Lower PE = cheaper
value_df["PE_Rank"] = (
    value_df["PE"]
    .rank(
        ascending=True,
        pct=True
    )
)

# Lower PB = cheaper
value_df["PB_Rank"] = (
    value_df["PB"]
    .rank(
        ascending=True,
        pct=True
    )
)

# Lower EV/EBITDA = cheaper
value_df["EV_EBITDA_Rank"] = (
    value_df["EV_EBITDA"]
    .rank(
        ascending=True,
        pct=True
    )
)

# Higher earnings yield = cheaper
value_df["EarningsYield_Rank"] = (
    value_df["EarningsYield"]
    .rank(
        ascending=False,
        pct=True
    )
)

# Higher FCF yield = cheaper
value_df["FCF_Yield_Rank"] = (
    value_df["FCF_Yield"]
    .rank(
        ascending=False,
        pct=True
    )
)


# ============================================================
# 5. COMPOSITE VALUE SCORE
# ============================================================

value_df["Value_Score"] = (

    0.25 * value_df["PE_Rank"]

    +

    0.20 * value_df["PB_Rank"]

    +

    0.20 * value_df["EV_EBITDA_Rank"]

    +

    0.20 * value_df["EarningsYield_Rank"]

    +

    0.15 * value_df["FCF_Yield_Rank"]

)


# Higher score = stronger value characteristics


# ============================================================
# 6. VALUE RANKING
# ============================================================

value_df["Value_Rank"] = (
    value_df["Value_Score"]
    .rank(
        ascending=False,
        pct=True
    )
)

value_df = value_df.sort_values(
    "Value_Score",
    ascending=False
).reset_index(drop=True)


print("\nValue Factor Ranking:")
display(
    value_df[
        [
            "Ticker",
            "Company",
            "PE",
            "PB",
            "EV_EBITDA",
            "EarningsYield",
            "FCF_Yield",
            "Value_Score"
        ]
    ]
)


# ============================================================
# 7. SELECT VALUE STOCKS
# ============================================================

TOP_N = 5

BOTTOM_N = 5

long_stocks = (
    value_df
    .head(TOP_N)
    ["Ticker"]
    .tolist()
)

short_stocks = (
    value_df
    .tail(BOTTOM_N)
    ["Ticker"]
    .tolist()
)

print("\nLong Value Portfolio:")
print(long_stocks)

print("\nShort Value Portfolio:")
print(short_stocks)


# ============================================================
# 8. HISTORICAL PRICES
# ============================================================

prices = yf.download(
    TICKERS,
    start=START_DATE,
    auto_adjust=True,
    progress=False
)

if isinstance(
    prices.columns,
    pd.MultiIndex
):

    close = prices["Close"]

else:

    close = prices[["Close"]]

    close.columns = [
        TICKERS[0]
    ]


close = close.ffill()

monthly_prices = (
    close
    .resample("ME")
    .last()
)

monthly_returns = (
    monthly_prices
    .pct_change()
)


# ============================================================
# 9. VALUE PORTFOLIO
# ============================================================

# Because current fundamental data is being used for
# the research ranking, the strategy is a current
# cross-sectional factor portfolio.

long_only_returns = (
    monthly_returns[
        [
            x for x in long_stocks
            if x in monthly_returns.columns
        ]
    ]
    .mean(axis=1)
)


# ============================================================
# 10. LONG-SHORT VALUE PORTFOLIO
# ============================================================

long_returns = (
    monthly_returns[
        [
            x for x in long_stocks
            if x in monthly_returns.columns
        ]
    ]
    .mean(axis=1)
)

short_returns = (
    monthly_returns[
        [
            x for x in short_stocks
            if x in monthly_returns.columns
        ]
    ]
    .mean(axis=1)
)

long_short_returns = (
    0.5 * long_returns
    -
    0.5 * short_returns
)


# ============================================================
# 11. NIFTY 50 BENCHMARK
# ============================================================

nifty = yf.download(
    "^NSEI",
    start=START_DATE,
    auto_adjust=True,
    progress=False
)["Close"]

nifty = nifty.squeeze()

nifty_monthly = (
    nifty
    .resample("ME")
    .last()
)

nifty_returns = (
    nifty_monthly
    .pct_change()
)


# ============================================================
# 12. COMBINE RESULTS
# ============================================================

results = pd.concat(

    [

        long_only_returns.rename(
            "Value Long Only"
        ),

        long_short_returns.rename(
            "Value Long Short"
        ),

        nifty_returns.rename(
            "NIFTY 50"
        )

    ],

    axis=1

).dropna()


# ============================================================
# 13. CUMULATIVE PERFORMANCE
# ============================================================

cumulative = (
    1 + results
).cumprod()


plt.figure(
    figsize=(15, 7)
)

for column in cumulative.columns:

    plt.plot(
        cumulative.index,
        cumulative[column],
        label=column
    )

plt.title(
    "Value Factor Strategy vs NIFTY 50"
)

plt.xlabel("Date")

plt.ylabel(
    "Growth of ₹1"
)

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()


# ============================================================
# 14. PERFORMANCE METRICS
# ============================================================

def calculate_metrics(
    returns
):

    returns = returns.dropna()

    equity = (
        1 + returns
    ).cumprod()

    cumulative_return = (
        equity.iloc[-1] - 1
    )

    years = (
        len(returns) / 12
    )

    annual_return = (
        equity.iloc[-1]
        ** (1 / years)
        - 1
        if years > 0
        else np.nan
    )

    annual_volatility = (
        returns.std()
        * np.sqrt(12)
    )

    sharpe = (

        returns.mean()
        /
        returns.std()
        * np.sqrt(12)

        if returns.std() > 0

        else np.nan
    )

    downside = returns[
        returns < 0
    ]

    sortino = (

        returns.mean()
        /
        downside.std()
        * np.sqrt(12)

        if len(downside) > 1
        and downside.std() > 0

        else np.nan
    )

    drawdown = (
        equity
        /
        equity.cummax()
        - 1
    )

    max_drawdown = (
        drawdown.min()
    )

    win_rate = (
        returns > 0
    ).mean()

    return {

        "Cumulative Return":
            cumulative_return,

        "Annual Return":
            annual_return,

        "Annual Volatility":
            annual_volatility,

        "Sharpe":
            sharpe,

        "Sortino":
            sortino,

        "Max Drawdown":
            max_drawdown,

        "Win Rate":
            win_rate
    }


metrics = []

for column in results.columns:

    row = calculate_metrics(
        results[column]
    )

    row["Strategy"] = column

    metrics.append(row)


metrics_df = (
    pd.DataFrame(metrics)
    .set_index("Strategy")
)


print(
    "\nPerformance Summary:"
)

display(
    metrics_df.style
    .format(
        {
            "Cumulative Return":
                "{:.2%}",

            "Annual Return":
                "{:.2%}",

            "Annual Volatility":
                "{:.2%}",

            "Max Drawdown":
                "{:.2%}",

            "Win Rate":
                "{:.2%}",

            "Sharpe":
                "{:.3f}",

            "Sortino":
                "{:.3f}"
        }
    )
)


# ============================================================
# 15. VALUE FACTOR EXCESS RETURN
# ============================================================

excess_return = (
    long_only_returns
    -
    nifty_returns
).dropna()

cumulative_excess = (
    1 + excess_return
).cumprod()


plt.figure(
    figsize=(15, 5)
)

plt.plot(
    cumulative_excess.index,
    cumulative_excess
)

plt.axhline(
    1,
    linestyle="--"
)

plt.title(
    "Value Factor Cumulative Excess Return"
)

plt.xlabel("Date")

plt.ylabel(
    "Growth of ₹1"
)

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()


# ============================================================
# 16. DRAWDOWN
# ============================================================

value_equity = (
    1 + long_only_returns
).cumprod()

value_drawdown = (
    value_equity
    /
    value_equity.cummax()
    - 1
)


plt.figure(
    figsize=(15, 5)
)

plt.fill_between(
    value_drawdown.index,
    value_drawdown.values,
    0
)

plt.title(
    "Value Factor Portfolio Drawdown"
)

plt.xlabel("Date")

plt.ylabel(
    "Drawdown"
)

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()


# ============================================================
# 17. FACTOR CHARACTERISTICS
# ============================================================

factor_characteristics = {

    "Average PE":
        value_df["PE"].mean(),

    "Average PB":
        value_df["PB"].mean(),

    "Average EV/EBITDA":
        value_df["EV_EBITDA"].mean(),

    "Average Earnings Yield":
        value_df["EarningsYield"].mean(),

    "Average FCF Yield":
        value_df["FCF_Yield"].mean()
}


print(
    "\nValue Factor Characteristics:"
)

for key, value in factor_characteristics.items():

    print(
        f"{key:25s}: {value:.4f}"
    )


# ============================================================
# 18. SECTOR DISTRIBUTION
# ============================================================

sector_distribution = (
    value_df
    .head(TOP_N)
    ["Sector"]
    .value_counts()
)


print(
    "\nTop Value Portfolio Sector Distribution:"
)

display(
    sector_distribution
)


plt.figure(
    figsize=(10, 5)
)

sector_distribution.plot(
    kind="bar"
)

plt.title(
    "Value Portfolio Sector Distribution"
)

plt.xlabel("Sector")

plt.ylabel("Number of Stocks")

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.show()


# ============================================================
# 19. ANNUAL RETURNS
# ============================================================

annual_returns = (
    long_only_returns
    .resample("YE")
    .apply(
        lambda x:
        (1 + x).prod() - 1
    )
)

print(
    "\nAnnual Value Factor Returns:"
)

display(
    annual_returns
    .to_frame("Return")
    .style
    .format("{:.2%}")
)


# ============================================================
# 20. SAVE RESEARCH OUTPUTS
# ============================================================

fundamental_df.to_csv(
    "Value_Factor_Fundamentals.csv",
    index=False
)

value_df.to_csv(
    "Value_Factor_Ranking.csv",
    index=False
)

results.to_csv(
    "Value_Factor_Strategy_Returns.csv"
)

metrics_df.to_csv(
    "Value_Factor_Performance.csv"
)

annual_returns.to_csv(
    "Value_Factor_Annual_Returns.csv"
)

print(
    "\n================================================"
)

print(
    "VALUE FACTOR RESEARCH COMPLETED"
)

print(
    "================================================"
)

print("\nFiles exported:")
print("• Value_Factor_Fundamentals.csv")
print("• Value_Factor_Ranking.csv")
print("• Value_Factor_Strategy_Returns.csv")
print("• Value_Factor_Performance.csv")
print("• Value_Factor_Annual_Returns.csv")
```
