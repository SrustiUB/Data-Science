```python
# ============================================================
# LOW-VOLATILITY FACTOR ALGORITHMIC TRADING
# ============================================================

!pip -q install yfinance

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import yfinance as yf


# ============================================================
# 1. STOCK UNIVERSE
# ============================================================

TICKERS = [
    "RELIANCE.NS",
    "HDFCBANK.NS",
    "ICICIBANK.NS",
    "INFY.NS",
    "TCS.NS",
    "ITC.NS",
    "BHARTIARTL.NS",
    "SBIN.NS",
    "LT.NS",
    "AXISBANK.NS",
    "KOTAKBANK.NS",
    "HINDUNILVR.NS",
    "MARUTI.NS",
    "SUNPHARMA.NS",
    "M&M.NS",
    "TATAMOTORS.NS",
    "ADANIENT.NS",
    "NTPC.NS",
    "POWERGRID.NS",
    "TATASTEEL.NS"
]

START_DATE = "2015-01-01"

print("Number of stocks:", len(TICKERS))


# ============================================================
# 2. GET STOCK INFORMATION
# ============================================================

stock_info = []

print("\nCollecting ticker information...\n")

for ticker in TICKERS:

    try:

        info = yf.Ticker(ticker).info

        stock_info.append({

            "Ticker": ticker,

            "Company":
                info.get(
                    "longName",
                    ticker
                ),

            "Sector":
                info.get(
                    "sector",
                    "N/A"
                ),

            "Industry":
                info.get(
                    "industry",
                    "N/A"
                ),

            "MarketCap":
                info.get(
                    "marketCap",
                    np.nan
                ),

            "Beta":
                info.get(
                    "beta",
                    np.nan
                ),

            "PE":
                info.get(
                    "trailingPE",
                    np.nan
                ),

            "PB":
                info.get(
                    "priceToBook",
                    np.nan
                )
        })

    except Exception:

        stock_info.append({

            "Ticker": ticker,
            "Company": ticker,
            "Sector": "N/A",
            "Industry": "N/A",
            "MarketCap": np.nan,
            "Beta": np.nan,
            "PE": np.nan,
            "PB": np.nan
        })


info_df = pd.DataFrame(
    stock_info
)

print("\nStock Information:")
display(info_df)


# ============================================================
# 3. DOWNLOAD HISTORICAL PRICE DATA
# ============================================================

prices = yf.download(
    TICKERS,
    start=START_DATE,
    auto_adjust=True,
    progress=False
)

if isinstance(
    prices.columns,
    pd.MultiIndex
):

    close = prices["Close"].copy()

else:

    close = prices[["Close"]].copy()

    close.columns = [
        TICKERS[0]
    ]


close = close.ffill()

print(
    "\nDaily price dataset:",
    close.shape
)


# ============================================================
# 4. DAILY RETURNS
# ============================================================

daily_returns = (
    close.pct_change()
)


# ============================================================
# 5. VOLATILITY MEASURES
# ============================================================

# 20-day realized volatility

vol_20 = (
    daily_returns
    .rolling(20)
    .std()
    * np.sqrt(252)
)

# 60-day realized volatility

vol_60 = (
    daily_returns
    .rolling(60)
    .std()
    * np.sqrt(252)
)

# 120-day realized volatility

vol_120 = (
    daily_returns
    .rolling(120)
    .std()
    * np.sqrt(252)
)

# Downside volatility

negative_returns = daily_returns.where(
    daily_returns < 0,
    0
)

downside_vol_60 = (
    negative_returns
    .rolling(60)
    .std()
    * np.sqrt(252)
)


# ============================================================
# 6. BETA CALCULATION
# ============================================================

nifty = yf.download(
    "^NSEI",
    start=START_DATE,
    auto_adjust=True,
    progress=False
)["Close"]

nifty = nifty.squeeze()

nifty_returns = (
    nifty.pct_change()
)


# Align benchmark

common_index = (
    daily_returns.index
    .intersection(
        nifty_returns.index
    )
)

daily_returns = (
    daily_returns
    .loc[common_index]
)

nifty_returns = (
    nifty_returns
    .loc[common_index]
)


# Rolling beta

rolling_beta = pd.DataFrame(
    index=daily_returns.index,
    columns=daily_returns.columns,
    dtype=float
)


for ticker in daily_returns.columns:

    covariance = (
        daily_returns[ticker]
        .rolling(60)
        .cov(nifty_returns)
    )

    benchmark_variance = (
        nifty_returns
        .rolling(60)
        .var()
    )

    rolling_beta[ticker] = (
        covariance /
        benchmark_variance
    )


# ============================================================
# 7. MONTH-END DATA
# ============================================================

monthly_prices = (
    close
    .resample("ME")
    .last()
)

monthly_returns = (
    monthly_prices
    .pct_change()
)


monthly_vol_20 = (
    vol_20
    .resample("ME")
    .last()
)

monthly_vol_60 = (
    vol_60
    .resample("ME")
    .last()
)

monthly_vol_120 = (
    vol_120
    .resample("ME")
    .last()
)

monthly_downside_vol = (
    downside_vol_60
    .resample("ME")
    .last()
)

monthly_beta = (
    rolling_beta
    .resample("ME")
    .last()
)


# ============================================================
# 8. LOW-VOLATILITY FACTOR SCORE
# ============================================================

# Lower volatility = better rank.
#
# percentile=True:
# Lower raw volatility gets a higher factor score.

vol20_rank = (
    monthly_vol_20
    .rank(
        axis=1,
        ascending=True,
        pct=True
    )
)

vol60_rank = (
    monthly_vol_60
    .rank(
        axis=1,
        ascending=True,
        pct=True
    )
)

vol120_rank = (
    monthly_vol_120
    .rank(
        axis=1,
        ascending=True,
        pct=True
    )
)

downside_rank = (
    monthly_downside_vol
    .rank(
        axis=1,
        ascending=True,
        pct=True
    )
)

beta_rank = (
    monthly_beta
    .rank(
        axis=1,
        ascending=True,
        pct=True
    )
)


# ============================================================
# 9. COMPOSITE LOW-VOLATILITY SCORE
# ============================================================

low_vol_score = (

    0.25 * vol20_rank

    +

    0.25 * vol60_rank

    +

    0.20 * vol120_rank

    +

    0.20 * downside_rank

    +

    0.10 * beta_rank

)


# ============================================================
# 10. CROSS-SECTIONAL RANK
# ============================================================

low_vol_rank = (
    low_vol_score
    .rank(
        axis=1,
        ascending=False,
        pct=True
    )
)


# ============================================================
# 11. LATEST FACTOR RANKING
# ============================================================

latest_score = pd.DataFrame({

    "Volatility_20D":
        monthly_vol_20.iloc[-1],

    "Volatility_60D":
        monthly_vol_60.iloc[-1],

    "Volatility_120D":
        monthly_vol_120.iloc[-1],

    "Downside_Volatility":
        monthly_downside_vol.iloc[-1],

    "Beta":
        monthly_beta.iloc[-1],

    "Low_Vol_Score":
        low_vol_score.iloc[-1],

    "Rank":
        low_vol_rank.iloc[-1]

})

latest_score = (
    latest_score
    .sort_values(
        "Low_Vol_Score",
        ascending=False
    )
)


print(
    "\nLatest Low-Volatility Ranking:"
)

display(
    latest_score
)


# ============================================================
# 12. PORTFOLIO SELECTION
# ============================================================

TOP_PERCENTILE = 0.80
BOTTOM_PERCENTILE = 0.20

low_vol_weights = pd.DataFrame(
    0.0,
    index=low_vol_rank.index,
    columns=low_vol_rank.columns
)

long_short_weights = pd.DataFrame(
    0.0,
    index=low_vol_rank.index,
    columns=low_vol_rank.columns
)


for date in low_vol_rank.index:

    ranks = low_vol_rank.loc[date]

    # --------------------------------------------------------
    # LOW-VOLATILITY LONG PORTFOLIO
    # --------------------------------------------------------

    long_stocks = ranks[
        ranks >= TOP_PERCENTILE
    ].dropna().index

    if len(long_stocks) > 0:

        low_vol_weights.loc[
            date,
            long_stocks
        ] = (
            1 /
            len(long_stocks)
        )

    # --------------------------------------------------------
    # LONG-SHORT PORTFOLIO
    # --------------------------------------------------------

    short_stocks = ranks[
        ranks <= BOTTOM_PERCENTILE
    ].dropna().index

    if len(long_stocks) > 0:

        long_short_weights.loc[
            date,
            long_stocks
        ] = (
            0.5 /
            len(long_stocks)
        )

    if len(short_stocks) > 0:

        long_short_weights.loc[
            date,
            short_stocks
        ] = (
            -0.5 /
            len(short_stocks)
        )


# ============================================================
# 13. PORTFOLIO RETURNS
# ============================================================

low_vol_returns = (
    low_vol_weights.shift(1)
    * monthly_returns
).sum(axis=1)

long_short_returns = (
    long_short_weights.shift(1)
    * monthly_returns
).sum(axis=1)


# ============================================================
# 14. NIFTY 50 MONTHLY BENCHMARK
# ============================================================

nifty_monthly = (
    nifty
    .resample("ME")
    .last()
)

nifty_monthly_returns = (
    nifty_monthly
    .pct_change()
)


# ============================================================
# 15. COMBINE RESULTS
# ============================================================

results = pd.concat(
    [

        low_vol_returns.rename(
            "Low Volatility Long Only"
        ),

        long_short_returns.rename(
            "Low Volatility Long Short"
        ),

        nifty_monthly_returns.rename(
            "NIFTY 50"
        )

    ],
    axis=1
).dropna()


# ============================================================
# 16. CUMULATIVE PERFORMANCE
# ============================================================

cumulative = (
    1 + results
).cumprod()


plt.figure(
    figsize=(15, 7)
)

for column in cumulative.columns:

    plt.plot(
        cumulative.index,
        cumulative[column],
        label=column
    )

plt.title(
    "Low-Volatility Factor Strategy vs NIFTY 50"
)

plt.xlabel("Date")
plt.ylabel("Growth of ₹1")

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()


# ============================================================
# 17. PERFORMANCE METRICS
# ============================================================

def calculate_metrics(returns):

    returns = returns.dropna()

    equity = (
        1 + returns
    ).cumprod()

    cumulative_return = (
        equity.iloc[-1] - 1
    )

    years = (
        len(returns) / 12
    )

    annual_return = (

        equity.iloc[-1]
        ** (1 / years)
        - 1

        if years > 0

        else np.nan
    )

    annual_volatility = (
        returns.std()
        * np.sqrt(12)
    )

    sharpe = (

        returns.mean()
        /
        returns.std()
        * np.sqrt(12)

        if returns.std() > 0

        else np.nan
    )

    downside = returns[
        returns < 0
    ]

    sortino = (

        returns.mean()
        /
        downside.std()
        * np.sqrt(12)

        if len(downside) > 1
        and downside.std() > 0

        else np.nan
    )

    drawdown = (
        equity
        /
        equity.cummax()
        - 1
    )

    max_drawdown = (
        drawdown.min()
    )

    win_rate = (
        returns > 0
    ).mean()

    return {

        "Cumulative Return":
            cumulative_return,

        "Annual Return":
            annual_return,

        "Annual Volatility":
            annual_volatility,

        "Sharpe":
            sharpe,

        "Sortino":
            sortino,

        "Max Drawdown":
            max_drawdown,

        "Win Rate":
            win_rate
    }


metrics = []

for column in results.columns:

    row = calculate_metrics(
        results[column]
    )

    row["Strategy"] = column

    metrics.append(row)


metrics_df = (
    pd.DataFrame(metrics)
    .set_index("Strategy")
)


print(
    "\nPerformance Summary:"
)

display(
    metrics_df.style
    .format(
        {
            "Cumulative Return":
                "{:.2%}",

            "Annual Return":
                "{:.2%}",

            "Annual Volatility":
                "{:.2%}",

            "Max Drawdown":
                "{:.2%}",

            "Win Rate":
                "{:.2%}",

            "Sharpe":
                "{:.3f}",

            "Sortino":
                "{:.3f}"
        }
    )
)


# ============================================================
# 18. LOW-VOLATILITY EXCESS RETURN
# ============================================================

excess_return = (
    low_vol_returns
    -
    nifty_monthly_returns
).dropna()

cumulative_excess = (
    1 + excess_return
).cumprod()


plt.figure(
    figsize=(15, 5)
)

plt.plot(
    cumulative_excess.index,
    cumulative_excess
)

plt.axhline(
    1,
    linestyle="--"
)

plt.title(
    "Low-Volatility Factor Cumulative Excess Return"
)

plt.xlabel("Date")
plt.ylabel("Growth of ₹1")

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()


# ============================================================
# 19. VOLATILITY COMPARISON
# ============================================================

portfolio_vol = (
    low_vol_returns
    .rolling(12)
    .std()
    * np.sqrt(12)
)

benchmark_vol = (
    nifty_monthly_returns
    .rolling(12)
    .std()
    * np.sqrt(12)
)

plt.figure(
    figsize=(15, 6)
)

plt.plot(
    portfolio_vol.index,
    portfolio_vol,
    label="Low-Volatility Portfolio"
)

plt.plot(
    benchmark_vol.index,
    benchmark_vol,
    label="NIFTY 50"
)

plt.title(
    "Rolling 12-Month Portfolio Volatility"
)

plt.xlabel("Date")
plt.ylabel("Annualized Volatility")

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()


# ============================================================
# 20. DRAWDOWN
# ============================================================

equity = (
    1 + low_vol_returns
).cumprod()

drawdown = (
    equity
    /
    equity.cummax()
    - 1
)


plt.figure(
    figsize=(15, 5)
)

plt.fill_between(
    drawdown.index,
    drawdown.values,
    0
)

plt.title(
    "Low-Volatility Portfolio Drawdown"
)

plt.xlabel("Date")
plt.ylabel("Drawdown")

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()


# ============================================================
# 21. ANNUAL RETURNS
# ============================================================

annual_returns = (
    low_vol_returns
    .resample("YE")
    .apply(
        lambda x:
        (1 + x).prod() - 1
    )
)

print(
    "\nAnnual Low-Volatility Factor Returns:"
)

display(
    annual_returns
    .to_frame("Return")
    .style
    .format("{:.2%}")
)


# ============================================================
# 22. PORTFOLIO TURNOVER
# ============================================================

turnover = (
    low_vol_weights
    .diff()
    .abs()
    .sum(axis=1)
)

print(
    "\nAverage Monthly Turnover:",
    f"{turnover.mean():.2%}"
)


# ============================================================
# 23. LATEST LOW-VOLATILITY PORTFOLIO
# ============================================================

latest_rank = latest_score.head(5)

latest_portfolio = latest_rank.index.tolist()

print(
    "\nTop 5 Low-Volatility Stocks:"
)

for i, ticker in enumerate(
    latest_portfolio,
    1
):

    print(
        f"{i}. {ticker}"
    )


# ============================================================
# 24. SECTOR DISTRIBUTION
# ============================================================

sector_map = (
    info_df
    .set_index("Ticker")
    ["Sector"]
)

latest_sectors = (
    sector_map
    .reindex(
        latest_portfolio
    )
)

sector_distribution = (
    latest_sectors
    .value_counts()
)

print(
    "\nLow-Volatility Portfolio Sector Distribution:"
)

display(
    sector_distribution
)


plt.figure(
    figsize=(10, 5)
)

sector_distribution.plot(
    kind="bar"
)

plt.title(
    "Low-Volatility Portfolio Sector Distribution"
)

plt.xlabel("Sector")
plt.ylabel("Number of Stocks")

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.show()


# ============================================================
# 25. FACTOR CHARACTERISTICS
# ============================================================

factor_characteristics = {

    "Average 20D Volatility":
        latest_score[
            "Volatility_20D"
        ].mean(),

    "Average 60D Volatility":
        latest_score[
            "Volatility_60D"
        ].mean(),

    "Average 120D Volatility":
        latest_score[
            "Volatility_120D"
        ].mean(),

    "Average Downside Volatility":
        latest_score[
            "Downside_Volatility"
        ].mean(),

    "Average Beta":
        latest_score[
            "Beta"
        ].mean()
}


print(
    "\nLow-Volatility Factor Characteristics:"
)

for key, value in factor_characteristics.items():

    print(
        f"{key:30s}: {value:.4f}"
    )


# ============================================================
# 26. SAVE RESEARCH OUTPUTS
# ============================================================

info_df.to_csv(
    "LowVol_Stock_Information.csv",
    index=False
)

monthly_vol_20.to_csv(
    "LowVol_20D_Volatility.csv"
)

monthly_vol_60.to_csv(
    "LowVol_60D_Volatility.csv"
)

monthly_vol_120.to_csv(
    "LowVol_120D_Volatility.csv"
)

monthly_beta.to_csv(
    "LowVol_Rolling_Beta.csv"
)

low_vol_score.to_csv(
    "LowVol_Factor_Score.csv"
)

low_vol_rank.to_csv(
    "LowVol_Factor_Rank.csv"
)

low_vol_weights.to_csv(
    "LowVol_Portfolio_Weights.csv"
)

results.to_csv(
    "LowVol_Strategy_Returns.csv"
)

metrics_df.to_csv(
    "LowVol_Performance.csv"
)

annual_returns.to_csv(
    "LowVol_Annual_Returns.csv"
)

sector_distribution.to_csv(
    "LowVol_Sector_Distribution.csv"
)

print(
    "\n================================================"
)

print(
    "LOW-VOLATILITY FACTOR RESEARCH COMPLETED"
)

print(
    "================================================"
)

print("\nFiles exported:")
print("• LowVol_Stock_Information.csv")
print("• LowVol_20D_Volatility.csv")
print("• LowVol_60D_Volatility.csv")
print("• LowVol_120D_Volatility.csv")
print("• LowVol_Rolling_Beta.csv")
print("• LowVol_Factor_Score.csv")
print("• LowVol_Factor_Rank.csv")
print("• LowVol_Portfolio_Weights.csv")
print("• LowVol_Strategy_Returns.csv")
print("• LowVol_Performance.csv")
print("• LowVol_Annual_Returns.csv")
print("• LowVol_Sector_Distribution.csv")
```
