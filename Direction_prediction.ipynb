```python
# ============================================================
# PROBABILISTIC NIFTY 50 DIRECTION — AutoML + QuantStats
# ============================================================

!pip -q install yfinance flaml quantstats

import yfinance as yf
import pandas as pd
import numpy as np
import quantstats as qs

from flaml import AutoML
from sklearn.metrics import (
    roc_auc_score, log_loss, brier_score_loss
)

# ------------------------------------------------------------
# 1. DATA
# ------------------------------------------------------------

df = yf.download(
    "^NSEI",
    start="2010-01-01",
    auto_adjust=False,
    progress=False
)

if isinstance(df.columns, pd.MultiIndex):
    df.columns = df.columns.get_level_values(0)

df["Return"] = df["Close"].pct_change()

# ------------------------------------------------------------
# 2. FEATURES
# ------------------------------------------------------------

for n in [1, 3, 5, 10, 20]:
    df[f"Ret_{n}"] = df["Close"].pct_change(n)

for n in [5, 10, 20]:
    df[f"Vol_{n}"] = df["Return"].rolling(n).std()

df["SMA20"] = df["Close"].rolling(20).mean()
df["SMA50"] = df["Close"].rolling(50).mean()
df["Dist_SMA20"] = df["Close"] / df["SMA20"] - 1
df["Dist_SMA50"] = df["Close"] / df["SMA50"] - 1

df["Range"] = (df["High"] - df["Low"]) / df["Close"]

# Target: positive next-day return
df["Target"] = (df["Return"].shift(-1) > 0).astype(int)

df = df.dropna()

features = [
    c for c in df.columns
    if c not in ["Open","High","Low","Close",
                 "Adj Close","Volume","Return","Target",
                 "SMA20","SMA50"]
]

X = df[features]
y = df["Target"]

# ------------------------------------------------------------
# 3. TIME-SERIES TRAIN / TEST
# ------------------------------------------------------------

split = int(len(df) * 0.8)

X_train, X_test = X.iloc[:split], X.iloc[split:]
y_train, y_test = y.iloc[:split], y.iloc[split:]

# ------------------------------------------------------------
# 4. AUTOML
# ------------------------------------------------------------

automl = AutoML()

automl.fit(
    X_train,
    y_train,
    task="classification",
    metric="log_loss",
    time_budget=60,
    seed=42
)

print("Best AutoML Model:")
print(automl.model)

# ------------------------------------------------------------
# 5. PROBABILITY PREDICTION
# ------------------------------------------------------------

prob = automl.predict_proba(X_test)[:, 1]

pred = (prob >= 0.50).astype(int)

# ------------------------------------------------------------
# 6. PROBABILISTIC METRICS
# ------------------------------------------------------------

print("\nProbability Metrics")
print("-------------------")

print("ROC-AUC :", round(
    roc_auc_score(y_test, prob), 4
))

print("Log Loss:", round(
    log_loss(y_test, prob), 4
))

print("Brier   :", round(
    brier_score_loss(y_test, prob), 4
))

# ------------------------------------------------------------
# 7. PROBABILITY-BASED STRATEGY
# ------------------------------------------------------------

test = df.iloc[split:].copy()

test["Probability"] = prob

# Long when probability > 60%
test["Position"] = np.where(
    test["Probability"] > 0.60, 1,
    np.where(test["Probability"] < 0.40, -1, 0)
)

# Signal available before next day's return
test["Position"] = test["Position"].shift(1).fillna(0)

test["Strategy_Return"] = (
    test["Position"] * test["Return"]
)

# ------------------------------------------------------------
# 8. QUANTSTATS
# ------------------------------------------------------------

strategy = test["Strategy_Return"]
benchmark = test["Return"]

print("\nQuantStats Performance")
print("----------------------")

print("Sharpe:",
      round(qs.stats.sharpe(strategy), 3))

print("CAGR:",
      round(qs.stats.cagr(strategy) * 100, 2), "%")

print("Max Drawdown:",
      round(qs.stats.max_drawdown(strategy) * 100, 2), "%")

print("Volatility:",
      round(qs.stats.volatility(strategy) * 100, 2), "%")

# Full QuantStats report
qs.reports.basic(strategy)

# ------------------------------------------------------------
# 9. LATEST PROBABILITY
# ------------------------------------------------------------

print("\nLatest NIFTY Probability")
print("------------------------")

print(
    "P(Next-Day Positive Return):",
    round(prob[-1] * 100, 2),
    "%"
)
```
