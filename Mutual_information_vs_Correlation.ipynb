```python

# Install libraries
!pip install yfinance scikit-learn --quiet

import yfinance as yf
import pandas as pd
import numpy as np
from sklearn.feature_selection import mutual_info_regression

# --------------------------
# NIFTY Stocks
# --------------------------
tickers = [
    "RELIANCE.NS","TCS.NS","INFY.NS","HDFCBANK.NS",
    "ICICIBANK.NS","ITC.NS","LT.NS","SBIN.NS",
    "HINDUNILVR.NS","BHARTIARTL.NS"
]

# --------------------------
# Download Data
# --------------------------
prices = yf.download(
    tickers,
    start="2020-01-01",
    end="2025-01-01",
    auto_adjust=True,
    progress=False
)["Close"]

prices = prices.dropna(axis=1)
returns = prices.pct_change().dropna()

# --------------------------
# Correlation Matrix
# --------------------------
corr_matrix = returns.corr()

print("\nCorrelation Matrix")
print(corr_matrix.round(2))

# --------------------------
# Mutual Information Matrix
# --------------------------
stocks = returns.columns
mi_matrix = pd.DataFrame(index=stocks, columns=stocks)

for i in stocks:
    for j in stocks:
        if i == j:
            mi_matrix.loc[i, j] = 1
        else:
            mi = mutual_info_regression(
                returns[[i]],
                returns[j],
                random_state=42
            )[0]
            mi_matrix.loc[i, j] = mi

mi_matrix = mi_matrix.astype(float)

print("\nMutual Information Matrix")
print(mi_matrix.round(3))

# --------------------------
# Average Dependency
# --------------------------
corr_score = corr_matrix.abs().mean().mean()
mi_score = mi_matrix.mean().mean()

print("\nAverage Absolute Correlation :", round(corr_score,3))
print("Average Mutual Information   :", round(mi_score,3))

# --------------------------
# Interpretation
# --------------------------
if mi_score > corr_score:
    print("\nMutual Information captured more dependency than correlation.")
else:
    print("\nCorrelation captured similar or stronger dependency.")
```
