```python
# ============================================================
# NIFTY MEAN-REVERSION PROBABILITY
# AutoML + QuantStats
# ============================================================

!pip -q install yfinance flaml quantstats

import yfinance as yf
import pandas as pd
import numpy as np
import quantstats as qs
from flaml import AutoML
from sklearn.metrics import roc_auc_score, log_loss, brier_score_loss

# 1. DATA
df = yf.download(
    "^NSEI",
    start="2010-01-01",
    auto_adjust=False,
    progress=False
)

if isinstance(df.columns, pd.MultiIndex):
    df.columns = df.columns.get_level_values(0)

df = df[["Open","High","Low","Close","Volume"]].copy()
df["Return"] = df["Close"].pct_change()

# 2. FEATURES
df["SMA20"] = df["Close"].rolling(20).mean()
df["STD20"] = df["Close"].rolling(20).std()

df["ZScore"] = (
    (df["Close"] - df["SMA20"]) /
    df["STD20"]
)

df["Distance"] = (
    df["Close"] / df["SMA20"] - 1
)

for n in [1, 5, 10, 20]:
    df[f"Ret_{n}"] = df["Close"].pct_change(n)

for n in [5, 10, 20]:
    df[f"Vol_{n}"] = df["Return"].rolling(n).std()

df["RSI"] = 100 - (
    100 /
    (
        1 +
        df["Return"].clip(lower=0).rolling(14).mean() /
        (-df["Return"].clip(upper=0)).rolling(14).mean()
    )
)

df["Range"] = (
    df["High"] - df["Low"]
) / df["Close"]

# 3. TARGET
# Mean-reversion = price moves back toward
# its 20-day mean within next 10 days

future_close = pd.concat(
    [df["Close"].shift(-i) for i in range(1,11)],
    axis=1
)

future_distance = abs(
    future_close.sub(df["SMA20"], axis=0)
    .div(df["SMA20"], axis=0)
)

df["Target"] = (
    future_distance.min(axis=1)
    < abs(df["Distance"]) * 0.50
).astype(int)

features = [
    "ZScore","Distance",
    "Ret_1","Ret_5","Ret_10","Ret_20",
    "Vol_5","Vol_10","Vol_20",
    "RSI","Range"
]

df = df.dropna(subset=features + ["Target"])

X = df[features]
y = df["Target"]

print("Dataset:", X.shape)
print("Mean-Reversion Rate:",
      round(y.mean()*100,2), "%")

# 4. TIME SPLIT
split = int(len(df)*0.80)

X_train, X_test = X.iloc[:split], X.iloc[split:]
y_train, y_test = y.iloc[:split], y.iloc[split:]

# 5. AUTOML
automl = AutoML()

automl.fit(
    X_train,
    y_train,
    task="classification",
    metric="log_loss",
    time_budget=60,
    seed=42,
    eval_method="holdout"
)

print("\nBest Model:")
print(automl.model)

# 6. PROBABILITY
prob = automl.predict_proba(X_test)[:,1]

print("\nProbability Metrics")
print("-------------------")
print("ROC-AUC:",
      round(roc_auc_score(y_test,prob),4))
print("Log Loss:",
      round(log_loss(y_test,prob),4))
print("Brier:",
      round(brier_score_loss(y_test,prob),4))

# 7. PROBABILITY-BASED STRATEGY
test = df.iloc[split:].copy()
test["Probability"] = prob

# High mean-reversion probability +
# negative Z-score -> long
# High probability +
# positive Z-score -> short

test["Position"] = np.where(
    (test["Probability"] >= 0.60) &
    (test["ZScore"] < -1),
    1,
    np.where(
        (test["Probability"] >= 0.60) &
        (test["ZScore"] > 1),
        -1,
        0
    )
)

test["Position"] = test["Position"].shift(1).fillna(0)

test["Strategy_Return"] = (
    test["Position"] * test["Return"]
)

# 8. QUANTSTATS
strategy = test["Strategy_Return"]

print("\nQuantStats")
print("----------")
print("Sharpe:",
      round(qs.stats.sharpe(strategy),3))

print("CAGR:",
      round(qs.stats.cagr(strategy)*100,2),"%")

print("Max Drawdown:",
      round(qs.stats.max_drawdown(strategy)*100,2),"%")

print("Volatility:",
      round(qs.stats.volatility(strategy)*100,2),"%")

qs.reports.basic(strategy)

# 9. LATEST PROBABILITY
print("\nLatest Mean-Reversion Probability:",
      round(prob[-1]*100,2), "%")

print("Current Z-Score:",
      round(test["ZScore"].iloc[-1],2))
```
